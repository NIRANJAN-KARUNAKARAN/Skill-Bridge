```python
def clean_job_text(text):

    text = safe_text(text)

    # Decode HTML entities
    text = html.unescape(text)

    # Remove HTML tags
    text = re.sub(r"<[^>]+>", " ", text)

    # Remove URLs
    text = re.sub(
        r"https?://\S+|www\.\S+",
        " ",
        text
    )

    # Normalize apostrophes
    text = text.replace("’", "'")
    text = text.replace("‘", "'")

    # Normalize whitespace
    text = re.sub(r"\s+", " ", text)

    return text.strip()


all_jobs["clean_text"] = (
    all_jobs["nlp_text"]
    .apply(clean_job_text)
)

display(
    all_jobs[
        ["jobId", "title", "clean_text"]
    ].head()
)
```

```python
texts = all_jobs["clean_text"].fillna("").astype(str).tolist()

docs = list(
    nlp.pipe(
        texts,
        batch_size=32
    )
)

print("Documents processed:", len(docs))
```

```python
def get_clean_tokens(doc):

    tokens = []

    for token in doc:

        if token.is_stop:
            continue

        if token.is_punct:
            continue

        if token.is_space:
            continue

        if not token.text.strip():
            continue

        token_text = token.text.lower().strip()

        if token_text:
            tokens.append(token_text)

    return tokens


all_jobs["clean_tokens"] = [
    get_clean_tokens(doc)
    for doc in docs
]

display(
    all_jobs[
        ["jobId", "title", "clean_tokens"]
    ].head()
)
```

```python
def get_lemmas(doc):

    lemmas = []

    for token in doc:

        if token.is_stop:
            continue

        if token.is_punct:
            continue

        if token.is_space:
            continue

        lemma = token.lemma_.lower().strip()

        if lemma:
            lemmas.append(lemma)

    return lemmas


all_jobs["lemmas"] = [
    get_lemmas(doc)
    for doc in docs
]

display(
    all_jobs[
        ["jobId", "title", "lemmas"]
    ].head()
)
```

```python
from spacy.matcher import PhraseMatcher

matcher = PhraseMatcher(
    nlp.vocab,
    attr="LOWER"
)

patterns = []

for skill in skill_vocabulary.keys():

    patterns.append(
        nlp.make_doc(skill)
    )

matcher.add(
    "SKILLS",
    patterns
)

skill_names = list(skill_vocabulary.keys())

print("Skill matcher ready.")
```

```python
def extract_skills(doc):

    matches = matcher(doc)

    found = []

    for match_id, start, end in matches:

        raw_skill = doc[start:end].text.lower().strip()

        normalized = skill_vocabulary.get(
            raw_skill,
            raw_skill
        )

        found.append(normalized)

    # Remove duplicates while preserving order
    unique_skills = list(
        dict.fromkeys(found)
    )

    return unique_skills


all_jobs["extracted_skills"] = [
    extract_skills(doc)
    for doc in docs
]

display(
    all_jobs[
        ["jobId", "title", "extracted_skills"]
    ].head(10)
)
```

```python
# ============================================================
# CELL 41 — MANUAL GOLD BENCHMARK CREATION
#              INDIA + MALAYSIA
# ============================================================

import pandas as pd
import json
import ast
import re
from google.colab import files

print("============================================")
print("MANUAL GOLD BENCHMARK CREATION")
print("============================================")

# ------------------------------------------------------------
# 1. CHECK REQUIRED DATASET
# ------------------------------------------------------------

if "all_jobs" not in globals():
    raise ValueError(
        "all_jobs was not found. Run the unified job dataset "
        "cells before running this cell."
    )

benchmark_source = all_jobs.copy()

print("Total jobs available:", len(benchmark_source))

# ------------------------------------------------------------
# 2. STANDARDIZE COUNTRY
# ------------------------------------------------------------

benchmark_source["country_clean"] = (
    benchmark_source["country"]
    .fillna("")
    .astype(str)
    .str.strip()
    .str.lower()
)

india = benchmark_source[
    benchmark_source["country_clean"].isin(
        ["india", "indian"]
    )
].copy()

malaysia = benchmark_source[
    benchmark_source["country_clean"].isin(
        ["malaysia", "malaysian"]
    )
].copy()

print("India jobs:", len(india))
print("Malaysia jobs:", len(malaysia))

# ------------------------------------------------------------
# 3. REQUIRE BOTH COUNTRIES
# ------------------------------------------------------------

if len(india) == 0 or len(malaysia) == 0:
    raise ValueError(
        "Both India and Malaysia records are required."
    )

# ------------------------------------------------------------
# 4. SELECT AT LEAST 100 JOB ADVERTISEMENTS
# ------------------------------------------------------------

TARGET_TOTAL = 100

india_target = min(50, len(india))
malaysia_target = min(50, len(malaysia))

# If one country has fewer than 50,
# redistribute remaining slots to the other country.

remaining = TARGET_TOTAL - (
    india_target + malaysia_target
)

if remaining > 0:

    india_extra = min(
        remaining,
        len(india) - india_target
    )

    india_target += india_extra
    remaining -= india_extra

if remaining > 0:

    malaysia_extra = min(
        remaining,
        len(malaysia) - malaysia_target
    )

    malaysia_target += malaysia_extra
    remaining -= malaysia_extra

actual_target = india_target + malaysia_target

if actual_target < TARGET_TOTAL:
    raise ValueError(
        f"Only {actual_target} jobs are available "
        f"for the benchmark. At least 100 are required."
    )

# Reproducible sampling
india_sample = india.sample(
    n=india_target,
    random_state=42
)

malaysia_sample = malaysia.sample(
    n=malaysia_target,
    random_state=42
)

benchmark = pd.concat(
    [india_sample, malaysia_sample],
    ignore_index=True
)

# ------------------------------------------------------------
# 5. KEEP IMPORTANT FIELDS
# ------------------------------------------------------------

required_columns = [
    "jobId",
    "title",
    "companyName",
    "country",
    "location",
    "jobDescription",
    "nlp_text"
]

available_columns = [
    col for col in required_columns
    if col in benchmark.columns
]

benchmark = benchmark[
    available_columns +
    [
        col for col in benchmark.columns
        if col not in available_columns
        and col not in ["country_clean"]
    ]
].copy()

# ------------------------------------------------------------
# 6. PARSE EXISTING AUTOMATIC SKILLS
# ------------------------------------------------------------

def parse_skill_list(value):

    if value is None:
        return []

    if isinstance(value, list):
        return [
            str(x).strip()
            for x in value
            if str(x).strip()
        ]

    if pd.isna(value):
        return []

    text = str(value).strip()

    if not text:
        return []

    # JSON list
    try:
        parsed = json.loads(text)

        if isinstance(parsed, list):
            return [
                str(x).strip()
                for x in parsed
                if str(x).strip()
            ]
    except:
        pass

    # Python list
    try:
        parsed = ast.literal_eval(text)

        if isinstance(parsed, list):
            return [
                str(x).strip()
                for x in parsed
                if str(x).strip()
            ]
    except:
        pass

    # Comma separated
    return [
        x.strip()
        for x in re.split(r",|;", text)
        if x.strip()
    ]


# ------------------------------------------------------------
# 7. PRE-FILL AUTOMATIC SKILLS
# ------------------------------------------------------------

if "normalized_skills" in benchmark.columns:

    benchmark["suggested_skills"] = (
        benchmark["normalized_skills"]
        .apply(parse_skill_list)
    )

elif "extracted_skills" in benchmark.columns:

    benchmark["suggested_skills"] = (
        benchmark["extracted_skills"]
        .apply(parse_skill_list)
    )

elif "tagsAndSkills" in benchmark.columns:

    benchmark["suggested_skills"] = (
        benchmark["tagsAndSkills"]
        .apply(parse_skill_list)
    )

else:

    benchmark["suggested_skills"] = [[] for _ in range(len(benchmark))]


# ------------------------------------------------------------
# 8. CREATE HUMAN VALIDATION COLUMNS
# ------------------------------------------------------------

# Human annotator will edit ONLY these fields.

benchmark["gold_skills"] = benchmark[
    "suggested_skills"
].apply(
    lambda x: ", ".join(x)
)

benchmark["validation_status"] = "REVIEW_REQUIRED"

benchmark["annotator_notes"] = ""

benchmark["skills_added_by_annotator"] = ""

benchmark["skills_removed_by_annotator"] = ""

# ------------------------------------------------------------
# 9. ANNOTATION INSTRUCTIONS
# ------------------------------------------------------------

benchmark["annotation_instruction"] = (
    "Review the job description and correct gold_skills. "
    "Add missing required skills and remove incorrect skills. "
    "Do not include generic words such as communication unless "
    "the job clearly treats them as a required skill."
)

# ------------------------------------------------------------
# 10. SORT BY COUNTRY
# ------------------------------------------------------------

benchmark = benchmark.sort_values(
    by=["country", "jobId"],
    na_position="last"
).reset_index(drop=True)

# ------------------------------------------------------------
# 11. SHOW SUMMARY
# ------------------------------------------------------------

print("\n============================================")
print("BENCHMARK CREATED")
print("============================================")

print("Benchmark records:", len(benchmark))

print("\nCountry distribution:")
print(
    benchmark["country"]
    .value_counts(dropna=False)
)

print("\nValidation status:")
print(
    benchmark["validation_status"]
    .value_counts()
)

# ------------------------------------------------------------
# 12. SAVE EXCEL FOR MANUAL REVIEW
# ------------------------------------------------------------

benchmark_output = (
    "04_MANUAL_Gold_Benchmark_India_Malaysia.xlsx"
)

benchmark.to_excel(
    benchmark_output,
    index=False
)

files.download(benchmark_output)

print("\nDownloaded:")
print(benchmark_output)

print("\nIMPORTANT:")
print("Open the downloaded Excel file.")
print("Review the gold_skills column manually.")
print("Change validation_status to VALIDATED after review.")
print("Then upload the reviewed file in the next cell.")
```

```python
# ============================================================
# CELL 43 — BASELINE SKILL EXTRACTION
#              DICTIONARY / REGEX METHOD
# ============================================================

import pandas as pd
import re
import json
from google.colab import files

print("============================================")
print("BASELINE SKILL EXTRACTION")
print("============================================")

# ------------------------------------------------------------
# 1. CHECK GOLD BENCHMARK
# ------------------------------------------------------------

if "validated_benchmark" not in globals():
    raise ValueError(
        "validated_benchmark not found. "
        "Run Cell 42 first."
    )

baseline_df = validated_benchmark.copy()

# ------------------------------------------------------------
# 2. BUILD SKILL VOCABULARY
# ------------------------------------------------------------

skill_vocabulary = set()

possible_skill_columns = [
    "normalized_skills",
    "extracted_skills",
    "tagsAndSkills",
    "gold_skills"
]

def parse_skills(value):

    if value is None:
        return []

    if isinstance(value, list):
        return [
            str(x).strip()
            for x in value
            if str(x).strip()
        ]

    if pd.isna(value):
        return []

    text = str(value).strip()

    if not text:
        return []

    # JSON
    try:
        parsed = json.loads(text)

        if isinstance(parsed, list):
            return [
                str(x).strip()
                for x in parsed
                if str(x).strip()
            ]
    except:
        pass

    # Python list
    try:
        parsed = eval(text)

        if isinstance(parsed, list):
            return [
                str(x).strip()
                for x in parsed
                if str(x).strip()
            ]
    except:
        pass

    return [
        x.strip()
        for x in re.split(r",|;", text)
        if x.strip()
    ]


for column in possible_skill_columns:

    if column in baseline_df.columns:

        for value in baseline_df[column]:

            skills = parse_skills(value)

            for skill in skills:

                if skill.strip():
                    skill_vocabulary.add(
                        skill.strip()
                    )

# ------------------------------------------------------------
# 3. ADD CYBERSECURITY VOCABULARY
# ------------------------------------------------------------

cybersecurity_vocabulary = [
    "Cybersecurity",
    "Information Security",
    "Network Security",
    "Application Security",
    "Cloud Security",
    "Security Operations",
    "Penetration Testing",
    "Ethical Hacking",
    "SIEM",
    "SOC",
    "Incident Response",
    "Vulnerability Assessment",
    "Risk Management",
    "IAM",
    "Identity and Access Management"
]

for skill in cybersecurity_vocabulary:
    skill_vocabulary.add(skill)

skill_vocabulary = sorted(
    skill_vocabulary,
    key=lambda x: len(x),
    reverse=True
)

print(
    "Baseline vocabulary size:",
    len(skill_vocabulary)
)

# ------------------------------------------------------------
# 4. PREPARE TEXT
# ------------------------------------------------------------

if "jobDescription" in baseline_df.columns:

    baseline_df["evaluation_text"] = (
        baseline_df["jobDescription"]
        .fillna("")
        .astype(str)
    )

elif "nlp_text" in baseline_df.columns:

    baseline_df["evaluation_text"] = (
        baseline_df["nlp_text"]
        .fillna("")
        .astype(str)
    )

else:

    raise ValueError(
        "No job description or nlp_text column found."
    )

# ------------------------------------------------------------
# 5. BASELINE EXTRACTION FUNCTION
# ------------------------------------------------------------

def baseline_extract(text):

    text = str(text).lower()

    detected = []

    for skill in skill_vocabulary:

        skill_lower = skill.lower()

        # Escape special regex characters
        pattern = (
            r"(?<![a-zA-Z0-9])"
            + re.escape(skill_lower)
            + r"(?![a-zA-Z0-9])"
        )

        if re.search(pattern, text):

            detected.append(skill)

    return list(dict.fromkeys(detected))


# ------------------------------------------------------------
# 6. RUN BASELINE
# ------------------------------------------------------------

print("\nRunning baseline extraction...")

baseline_df["baseline_skills"] = (
    baseline_df["evaluation_text"]
    .apply(baseline_extract)
)

# ------------------------------------------------------------
# 7. DISPLAY RESULTS
# ------------------------------------------------------------

baseline_df["baseline_skill_count"] = (
    baseline_df["baseline_skills"]
    .apply(len)
)

print("\n============================================")
print("BASELINE RESULTS")
print("============================================")

print(
    "Advertisements processed:",
    len(baseline_df)
)

print(
    "Advertisements with extracted skills:",
    (
        baseline_df["baseline_skill_count"] > 0
    ).sum()
)

print(
    "Advertisements with zero extracted skills:",
    (
        baseline_df["baseline_skill_count"] == 0
    ).sum()
)

print(
    "Average skills per advertisement:",
    round(
        baseline_df["baseline_skill_count"].mean(),
        2
    )
)

display(
    baseline_df[
        [
            "jobId",
            "title",
            "country",
            "baseline_skills",
            "gold_skills"
        ]
    ].head(10)
)

# ------------------------------------------------------------
# 8. SAVE BASELINE OUTPUT
# ------------------------------------------------------------

baseline_export = baseline_df.copy()

baseline_export["baseline_skills"] = (
    baseline_export["baseline_skills"]
    .apply(
        lambda x: json.dumps(
            x,
            ensure_ascii=False
        )
    )
)

baseline_output = (
    "21_Benchmark_Baseline_Extraction.xlsx"
)

baseline_export.to_excel(
    baseline_output,
    index=False
)

files.download(baseline_output)

print("\nDownloaded:")
print(baseline_output)
```

```python
# ============================================================
# CELL 44 — BASELINE PRECISION, RECALL & F1
#              OVERALL + COUNTRY + ROLE
# ============================================================

import pandas as pd
import json
import ast
from google.colab import files

print("============================================")
print("BASELINE EVALUATION")
print("PRECISION / RECALL / F1")
print("============================================")

# ------------------------------------------------------------
# 1. CHECK PREVIOUS CELL
# ------------------------------------------------------------

if "baseline_df" not in globals():
    raise ValueError(
        "baseline_df not found. Run Cell 43 first."
    )

evaluation_df = baseline_df.copy()

# ------------------------------------------------------------
# 2. SKILL PARSER
# ------------------------------------------------------------

def parse_skill_list(value):

    if value is None:
        return []

    if isinstance(value, list):
        return [
            str(x).strip().lower()
            for x in value
            if str(x).strip()
        ]

    if pd.isna(value):
        return []

    text = str(value).strip()

    if not text:
        return []

    # JSON
    try:
        parsed = json.loads(text)

        if isinstance(parsed, list):
            return [
                str(x).strip().lower()
                for x in parsed
                if str(x).strip()
            ]
    except:
        pass

    # Python list
    try:
        parsed = ast.literal_eval(text)

        if isinstance(parsed, list):
            return [
                str(x).strip().lower()
                for x in parsed
                if str(x).strip()
            ]
    except:
        pass

    # Comma / semicolon separated
    return [
        x.strip().lower()
        for x in text.split(",")
        if x.strip()
    ]


# ------------------------------------------------------------
# 3. CREATE GOLD SKILL LIST
# ------------------------------------------------------------

evaluation_df["gold_skill_list"] = (
    evaluation_df["gold_skills"]
    .apply(parse_skill_list)
)

evaluation_df["baseline_skill_list"] = (
    evaluation_df["baseline_skills"]
    .apply(parse_skill_list)
)

# ------------------------------------------------------------
# 4. CALCULATE TP / FP / FN
# ------------------------------------------------------------

def calculate_counts(gold, predicted):

    gold_set = set(gold)
    predicted_set = set(predicted)

    tp = len(
        gold_set.intersection(predicted_set)
    )

    fp = len(
        predicted_set - gold_set
    )

    fn = len(
        gold_set - predicted_set
    )

    return tp, fp, fn


results = []

for _, row in evaluation_df.iterrows():

    tp, fp, fn = calculate_counts(
        row["gold_skill_list"],
        row["baseline_skill_list"]
    )

    results.append(
        {
            "jobId": row["jobId"],
            "title": row["title"],
            "country": row["country"],
            "TP": tp,
            "FP": fp,
            "FN": fn
        }
    )

counts_df = pd.DataFrame(results)

# ------------------------------------------------------------
# 5. METRIC FUNCTION
# ------------------------------------------------------------

def calculate_metrics(tp, fp, fn):

    precision = (
        tp / (tp + fp)
        if (tp + fp) > 0
        else 0
    )

    recall = (
        tp / (tp + fn)
        if (tp + fn) > 0
        else 0
    )

    f1 = (
        2 * precision * recall /
        (precision + recall)
        if (precision + recall) > 0
        else 0
    )

    return precision, recall, f1


# ------------------------------------------------------------
# 6. OVERALL METRICS
# ------------------------------------------------------------

overall_tp = counts_df["TP"].sum()
overall_fp = counts_df["FP"].sum()
overall_fn = counts_df["FN"].sum()

precision, recall, f1 = calculate_metrics(
    overall_tp,
    overall_fp,
    overall_fn
)

overall_results = pd.DataFrame(
    [
        {
            "Method": "Dictionary / Regex Baseline",
            "TP": overall_tp,
            "FP": overall_fp,
            "FN": overall_fn,
            "Precision": round(precision, 4),
            "Recall": round(recall, 4),
            "F1": round(f1, 4)
        }
    ]
)

print("\n============================================")
print("OVERALL RESULTS")
print("============================================")

display(overall_results)

# ------------------------------------------------------------
# 7. COUNTRY METRICS
# ------------------------------------------------------------

country_results = []

for country, group in counts_df.groupby(
    "country",
    dropna=False
):

    tp = group["TP"].sum()
    fp = group["FP"].sum()
    fn = group["FN"].sum()

    p, r, f = calculate_metrics(
        tp, fp, fn
    )

    country_results.append(
        {
            "Country": country,
            "Advertisements": len(group),
            "TP": tp,
            "FP": fp,
            "FN": fn,
            "Precision": round(p, 4),
            "Recall": round(r, 4),
            "F1": round(f, 4)
        }
    )

country_results_df = pd.DataFrame(
    country_results
)

print("\n============================================")
print("RESULTS BY COUNTRY")
print("============================================")

display(country_results_df)

# ------------------------------------------------------------
# 8. ROLE / JOB TITLE METRICS
# ------------------------------------------------------------

role_results = []

for role, group in counts_df.groupby(
    "title",
    dropna=False
):

    tp = group["TP"].sum()
    fp = group["FP"].sum()
    fn = group["FN"].sum()

    p, r, f = calculate_metrics(
        tp, fp, fn
    )

    role_results.append(
        {
            "Role / Job Title": role,
            "Advertisements": len(group),
            "TP": tp,
            "FP": fp,
            "FN": fn,
            "Precision": round(p, 4),
            "Recall": round(r, 4),
            "F1": round(f, 4)
        }
    )

role_results_df = pd.DataFrame(
    role_results
)

print("\n============================================")
print("RESULTS BY ROLE / JOB TITLE")
print("============================================")

display(
    role_results_df.head(30)
)

print(
    "\nTotal distinct job titles evaluated:",
    len(role_results_df)
)

# ------------------------------------------------------------
# 9. ZERO-SKILL BASELINE RESULTS
# ------------------------------------------------------------

zero_skill_count = (
    evaluation_df["baseline_skill_list"]
    .apply(len)
    .eq(0)
    .sum()
)

zero_skill_rate = (
    zero_skill_count /
    len(evaluation_df)
    if len(evaluation_df) > 0
    else 0
)

zero_skill_summary = pd.DataFrame(
    [
        {
            "Total Advertisements": len(evaluation_df),
            "Zero Skill Predictions": zero_skill_count,
            "Zero Skill Rate": round(
                zero_skill_rate,
                4
            )
        }
    ]
)

print("\n============================================")
print("ZERO-SKILL ANALYSIS")
print("============================================")

display(zero_skill_summary)

# ------------------------------------------------------------
# 10. COMBINE RESULTS
# ------------------------------------------------------------

evaluation_results = {
    "overall": overall_results,
    "country": country_results_df,
    "role": role_results_df,
    "zero_skill": zero_skill_summary,
    "record_level_counts": counts_df
}

# ------------------------------------------------------------
# 11. DOWNLOAD RESULTS
# ------------------------------------------------------------

baseline_metrics_output = (
    "22_Baseline_Precision_Recall_F1.xlsx"
)

with pd.ExcelWriter(
    baseline_metrics_output,
    engine="openpyxl"
) as writer:

    overall_results.to_excel(
        writer,
        sheet_name="Overall",
        index=False
    )

    country_results_df.to_excel(
        writer,
        sheet_name="By_Country",
        index=False
    )

    role_results_df.to_excel(
        writer,
        sheet_name="By_Role",
        index=False
    )

    zero_skill_summary.to_excel(
        writer,
        sheet_name="Zero_Skill",
        index=False
    )

    counts_df.to_excel(
        writer,
        sheet_name="Record_Level",
        index=False
    )

files.download(
    baseline_metrics_output
)

print("\nDownloaded:")
print(baseline_metrics_output)
```

```python
# ============================================================
# CELL 45 — STRONG NLP / SEMANTIC SKILL EXTRACTION
#              SENTENCE TRANSFORMER
# ============================================================

!pip -q install sentence-transformers

import pandas as pd
import numpy as np
import json
from sentence_transformers import SentenceTransformer, util
from google.colab import files

print("============================================")
print("STRONG NLP / SEMANTIC SKILL EXTRACTION")
print("============================================")

# ------------------------------------------------------------
# 1. CHECK PREVIOUS CELL
# ------------------------------------------------------------

if "evaluation_df" not in globals():
    raise ValueError(
        "evaluation_df not found. Run Cell 44 first."
    )

semantic_df = evaluation_df.copy()

# ------------------------------------------------------------
# 2. PREPARE TEXT
# ------------------------------------------------------------

if "jobDescription" in semantic_df.columns:

    semantic_df["semantic_text"] = (
        semantic_df["jobDescription"]
        .fillna("")
        .astype(str)
        .str.strip()
    )

elif "nlp_text" in semantic_df.columns:

    semantic_df["semantic_text"] = (
        semantic_df["nlp_text"]
        .fillna("")
        .astype(str)
        .str.strip()
    )

else:

    raise ValueError(
        "No jobDescription or nlp_text column found."
    )

# ------------------------------------------------------------
# 3. BUILD SKILL VOCABULARY
# ------------------------------------------------------------

skill_vocabulary = set()

for column in [
    "gold_skills",
    "normalized_skills",
    "extracted_skills",
    "tagsAndSkills"
]:

    if column not in semantic_df.columns:
        continue

    for value in semantic_df[column]:

        if value is None or pd.isna(value):
            continue

        if isinstance(value, list):

            for skill in value:

                skill = str(skill).strip()

                if skill:
                    skill_vocabulary.add(skill)

        else:

            text = str(value).strip()

            if not text:
                continue

            try:

                parsed = json.loads(text)

                if isinstance(parsed, list):

                    for skill in parsed:

                        skill = str(skill).strip()

                        if skill:
                            skill_vocabulary.add(skill)

                    continue

            except:
                pass

            for skill in text.replace(
                ";", ","
            ).split(","):

                skill = skill.strip()

                if skill:
                    skill_vocabulary.add(skill)

# ------------------------------------------------------------
# 4. CYBERSECURITY VOCABULARY
# ------------------------------------------------------------

cybersecurity_vocabulary = [
    "Cybersecurity",
    "Information Security",
    "Network Security",
    "Application Security",
    "Cloud Security",
    "Security Operations",
    "Penetration Testing",
    "Ethical Hacking",
    "SIEM",
    "SOC",
    "Incident Response",
    "Vulnerability Assessment",
    "Risk Management",
    "IAM",
    "Identity and Access Management"
]

skill_vocabulary.update(
    cybersecurity_vocabulary
)

skill_vocabulary = sorted(
    skill_vocabulary
)

print(
    "Semantic skill vocabulary:",
    len(skill_vocabulary)
)

# ------------------------------------------------------------
# 5. LOAD SENTENCE TRANSFORMER
# ------------------------------------------------------------

model_name = "all-MiniLM-L6-v2"

model = SentenceTransformer(
    model_name
)

print(
    "Loaded model:",
    model_name
)

# ------------------------------------------------------------
# 6. CREATE SKILL EMBEDDINGS ONCE
# ------------------------------------------------------------

print("\nCreating skill embeddings...")

skill_embeddings = model.encode(
    skill_vocabulary,
    convert_to_tensor=True,
    normalize_embeddings=True,
    show_progress_bar=True
)

# ------------------------------------------------------------
# 7. BATCH ENCODE JOB DESCRIPTIONS
# ------------------------------------------------------------

texts = (
    semantic_df["semantic_text"]
    .tolist()
)

print(
    "\nEncoding",
    len(texts),
    "job advertisements..."
)

text_embeddings = model.encode(
    texts,
    batch_size=64,
    convert_to_tensor=True,
    normalize_embeddings=True,
    show_progress_bar=True
)

# ------------------------------------------------------------
# 8. CALCULATE SIMILARITIES
# ------------------------------------------------------------

print("\nCalculating semantic similarities...")

similarity_matrix = util.cos_sim(
    text_embeddings,
    skill_embeddings
)

# ------------------------------------------------------------
# 9. EXTRACT SEMANTIC SKILLS
# ------------------------------------------------------------

THRESHOLD = 0.55
TOP_K = 20

semantic_skills = []
semantic_confidence = []

for row_index in range(
    similarity_matrix.shape[0]
):

    similarities = similarity_matrix[
        row_index
    ]

    top_k = min(
        TOP_K,
        len(skill_vocabulary)
    )

    top_indices = similarities.topk(
        k=top_k
    ).indices.tolist()

    detected = []
    confidence_items = []

    for idx in top_indices:

        score = float(
            similarities[idx]
        )

        if score >= THRESHOLD:

            skill = skill_vocabulary[
                int(idx)
            ]

            detected.append(skill)

            confidence_items.append(
                {
                    "skill": skill,
                    "confidence": round(
                        score,
                        4
                    )
                }
            )

    # Remove duplicates
    unique_detected = list(
        dict.fromkeys(detected)
    )

    semantic_skills.append(
        unique_detected
    )

    semantic_confidence.append(
        confidence_items
    )

# ------------------------------------------------------------
# 10. STORE RESULTS
# ------------------------------------------------------------

semantic_df["semantic_skills"] = (
    semantic_skills
)

semantic_df["semantic_skill_count"] = (
    semantic_df["semantic_skills"]
    .apply(len)
)

semantic_df["semantic_confidence"] = (
    semantic_confidence
)

def average_confidence(values):

    if not values:
        return None

    scores = [
        item["confidence"]
        for item in values
        if isinstance(item, dict)
        and "confidence" in item
    ]

    if not scores:
        return None

    return round(
        float(np.mean(scores)),
        4
    )

semantic_df["semantic_confidence_score"] = (
    semantic_df["semantic_confidence"]
    .apply(average_confidence)
)

semantic_df["semantic_model"] = model_name

semantic_df["semantic_threshold"] = (
    THRESHOLD
)

# ------------------------------------------------------------
# 11. RESULTS SUMMARY
# ------------------------------------------------------------

zero_semantic = (
    semantic_df["semantic_skill_count"]
    .eq(0)
    .sum()
)

with_semantic = (
    semantic_df["semantic_skill_count"]
    .gt(0)
    .sum()
)

print("\n============================================")
print("SEMANTIC EXTRACTION RESULTS")
print("============================================")

print(
    "Advertisements processed:",
    len(semantic_df)
)

print(
    "With semantic skills:",
    with_semantic
)

print(
    "Without semantic skills:",
    zero_semantic
)

print(
    "Average semantic skills:",
    round(
        semantic_df[
            "semantic_skill_count"
        ].mean(),
        2
    )
)

# ------------------------------------------------------------
# 12. PREVIEW
# ------------------------------------------------------------

display(
    semantic_df[
        [
            "jobId",
            "title",
            "country",
            "gold_skills",
            "baseline_skills",
            "semantic_skills",
            "semantic_confidence_score"
        ]
    ].head(10)
)

# ------------------------------------------------------------
# 13. EXPORT
# ------------------------------------------------------------

semantic_export = semantic_df.copy()

for column in [
    "baseline_skills",
    "gold_skill_list",
    "semantic_skills",
    "semantic_confidence"
]:

    if column in semantic_export.columns:

        semantic_export[column] = (
            semantic_export[column]
            .apply(
                lambda x: json.dumps(
                    x,
                    ensure_ascii=False
                )
            )
        )

semantic_output = (
    "23_Semantic_Skill_Extraction.xlsx"
)

semantic_export.to_excel(
    semantic_output,
    index=False
)

files.download(
    semantic_output
)

print("\nDownloaded:")
print(semantic_output)
```

```python
# ============================================================
# CELL 47 — SECOND STRONGER NLP METHOD
# Transformer-based Skill Extraction
# ============================================================

!pip -q install transformers torch

import pandas as pd
import numpy as np
import re
import ast
import torch

from transformers import pipeline


# ------------------------------------------------------------
# 1. Select dataframe
# ------------------------------------------------------------
if "semantic_df" in globals():
    transformer_df = semantic_df.copy()
elif "evaluation_df" in globals():
    transformer_df = evaluation_df.copy()
else:
    raise ValueError(
        "❌ Benchmark dataframe not found. Run previous cells first."
    )


# ------------------------------------------------------------
# 2. Find job description column
# ------------------------------------------------------------
if "jobDescription" in transformer_df.columns:
    text_column = "jobDescription"

elif "job_description" in transformer_df.columns:
    text_column = "job_description"

elif "nlp_text" in transformer_df.columns:
    text_column = "nlp_text"

else:
    raise ValueError(
        "❌ Job description column not found."
    )


# ------------------------------------------------------------
# 3. Skill parser
# ------------------------------------------------------------
def parse_skill_values(value):

    if value is None:
        return []

    if isinstance(value, float) and pd.isna(value):
        return []

    if isinstance(value, (list, tuple, set)):
        return [
            str(x).strip()
            for x in value
            if str(x).strip()
        ]

    text = str(value).strip()

    if not text:
        return []

    try:
        parsed = ast.literal_eval(text)

        if isinstance(parsed, (list, tuple, set)):
            return [
                str(x).strip()
                for x in parsed
                if str(x).strip()
            ]
    except:
        pass

    return [
        x.strip()
        for x in text.split(",")
        if x.strip()
    ]


# ------------------------------------------------------------
# 4. Build skill vocabulary
# ------------------------------------------------------------
skill_vocabulary = set()

source_columns = [
    "gold_skills",
    "baseline_skills",
    "semantic_skills",
    "normalized_skills",
    "extracted_skills",
    "tagsAndSkills"
]

for column in source_columns:

    if column in transformer_df.columns:

        for value in transformer_df[column]:

            for skill in parse_skill_values(value):

                skill = skill.strip().lower()

                if len(skill) >= 2:
                    skill_vocabulary.add(skill)


# Cybersecurity vocabulary
cybersecurity_terms = {
    "cybersecurity",
    "cyber security",
    "information security",
    "infosec",
    "network security",
    "application security",
    "appsec",
    "cloud security",
    "security operations",
    "soc",
    "siem",
    "penetration testing",
    "penetration test",
    "vulnerability assessment",
    "incident response",
    "digital forensics",
    "firewall",
    "ids",
    "ips",
    "iam",
    "identity and access management",
    "zero trust"
}

skill_vocabulary.update(cybersecurity_terms)

skill_vocabulary = sorted(skill_vocabulary)

print("Total skill vocabulary:", len(skill_vocabulary))


# ------------------------------------------------------------
# 5. Load transformer NER model
# ------------------------------------------------------------
print("\nLoading transformer model...")

ner_model = pipeline(
    "token-classification",
    model="dslim/bert-base-NER",
    aggregation_strategy="simple"
)

print("✅ Transformer model loaded.")


# ------------------------------------------------------------
# 6. Transformer extraction
# ------------------------------------------------------------
def transformer_extract(text):

    if pd.isna(text):
        return []

    text = str(text).strip()

    if not text:
        return []

    # Limit text length to avoid excessive processing
    text = text[:4000]

    try:
        entities = ner_model(text)
    except Exception:
        return []

    extracted = []

    for entity in entities:

        word = entity.get("word", "").strip()

        if not word:
            continue

        # Keep useful entity phrases
        word = re.sub(r"\s+", " ", word).strip()

        if len(word) >= 2:
            extracted.append(word.lower())

    return list(dict.fromkeys(extracted))


# ------------------------------------------------------------
# 7. Map transformer entities to known skills
# ------------------------------------------------------------
def map_to_skills(entities, vocabulary):

    results = []

    for entity in entities:

        entity_clean = entity.lower().strip()

        # Exact match
        if entity_clean in vocabulary:
            results.append(entity_clean)
            continue

        # Partial phrase matching
        for skill in vocabulary:

            if (
                len(skill) >= 3
                and (
                    skill in entity_clean
                    or entity_clean in skill
                )
            ):
                results.append(skill)

    return list(dict.fromkeys(results))


# ------------------------------------------------------------
# 8. Process benchmark
# ------------------------------------------------------------
transformer_entities = []
transformer_skills = []

print("\nRunning transformer extraction...")

for i, text in enumerate(transformer_df[text_column]):

    entities = transformer_extract(text)

    skills = map_to_skills(
        entities,
        skill_vocabulary
    )

    transformer_entities.append(entities)
    transformer_skills.append(skills)

    if (i + 1) % 10 == 0:
        print(
            f"Processed {i + 1}/{len(transformer_df)}"
        )


# ------------------------------------------------------------
# 9. Store results
# ------------------------------------------------------------
transformer_df["transformer_entities"] = transformer_entities

transformer_df["transformer_skills"] = transformer_skills

transformer_df["transformer_model"] = (
    "dslim/bert-base-NER"
)

transformer_df["transformer_method"] = (
    "Transformer Token Classification + Skill Mapping"
)


# ------------------------------------------------------------
# 10. Zero-skill count
# ------------------------------------------------------------
transformer_df["transformer_skill_count"] = (
    transformer_df["transformer_skills"]
    .apply(len)
)

zero_count = int(
    (
        transformer_df["transformer_skill_count"] == 0
    ).sum()
)

print("\n==========================================")
print("TRANSFORMER EXTRACTION COMPLETED")
print("==========================================")

print(
    "Total benchmark records:",
    len(transformer_df)
)

print(
    "Records with extracted skills:",
    len(transformer_df) - zero_count
)

print(
    "Records with zero skills:",
    zero_count
)


# ------------------------------------------------------------
# 11. Save output
# ------------------------------------------------------------
output_file = (
    "25_Transformer_Skill_Extraction.xlsx"
)

with pd.ExcelWriter(
    output_file,
    engine="openpyxl"
) as writer:

    transformer_df.to_excel(
        writer,
        sheet_name="Transformer_Results",
        index=False
    )

    pd.DataFrame({
        "Metric": [
            "Total Records",
            "Records With Skills",
            "Zero Skill Records",
            "Zero Skill Percentage"
        ],
        "Value": [
            len(transformer_df),
            len(transformer_df) - zero_count,
            zero_count,
            round(
                zero_count / len(transformer_df) * 100,
                2
            ) if len(transformer_df) else 0
        ]
    }).to_excel(
        writer,
        sheet_name="Summary",
        index=False
    )


# ------------------------------------------------------------
# 12. Download
# ------------------------------------------------------------
from google.colab import files

files.download(output_file)

print("\n✅ Cell 47 completed.")
print("📥 Download:", output_file)
```

```python
# ============================================================
# CELL 48 — FINAL COMPARISON OF 3 EXTRACTION METHODS
# ============================================================

import pandas as pd
import numpy as np
import ast

# ------------------------------------------------------------
# 1. Check dataframe
# ------------------------------------------------------------
if "transformer_df" not in globals():
    raise ValueError(
        "❌ transformer_df not found. Please run Cell 47 first."
    )

df = transformer_df.copy()

print("Total benchmark records:", len(df))


# ------------------------------------------------------------
# 2. Skill parsing function
# ------------------------------------------------------------
def parse_skills(value):

    if value is None:
        return set()

    if isinstance(value, float) and pd.isna(value):
        return set()

    if isinstance(value, (list, tuple, set)):
        items = value

    else:
        text = str(value).strip()

        if not text or text.lower() in ["nan", "none", "[]"]:
            return set()

        try:
            parsed = ast.literal_eval(text)

            if isinstance(parsed, (list, tuple, set)):
                items = parsed
            else:
                items = text.split(",")

        except:
            items = text.split(",")

    return {
        str(x).strip().lower()
        for x in items
        if str(x).strip()
    }


# ------------------------------------------------------------
# 3. Evaluation function
# ------------------------------------------------------------
def evaluate(data, prediction_column):

    TP = 0
    FP = 0
    FN = 0

    for _, row in data.iterrows():

        gold = parse_skills(row["gold_skills"])
        predicted = parse_skills(row[prediction_column])

        TP += len(gold & predicted)
        FP += len(predicted - gold)
        FN += len(gold - predicted)

    precision = (
        TP / (TP + FP)
        if (TP + FP) > 0 else 0
    )

    recall = (
        TP / (TP + FN)
        if (TP + FN) > 0 else 0
    )

    f1 = (
        2 * precision * recall / (precision + recall)
        if (precision + recall) > 0 else 0
    )

    return {
        "TP": TP,
        "FP": FP,
        "FN": FN,
        "Precision": round(precision, 4),
        "Recall": round(recall, 4),
        "F1": round(f1, 4)
    }


# ------------------------------------------------------------
# 4. Define methods
# ------------------------------------------------------------
methods = {
    "Baseline Dictionary/Regex": "baseline_skills",
    "Sentence Transformer": "semantic_skills",
    "Transformer NER": "transformer_skills"
}


# ------------------------------------------------------------
# 5. Overall comparison
# ------------------------------------------------------------
overall_rows = []

for method_name, column in methods.items():

    if column not in df.columns:
        print(f"⚠️ Skipping {method_name}: {column} not found")
        continue

    result = evaluate(df, column)

    overall_rows.append({
        "Method": method_name,
        "Records": len(df),
        **result
    })

overall_results = pd.DataFrame(overall_rows)

print("\n==============================================")
print("OVERALL COMPARISON")
print("==============================================")

display(overall_results)


# ------------------------------------------------------------
# 6. Country comparison
# ------------------------------------------------------------
country_rows = []

if "country" in df.columns:

    for country in sorted(
        df["country"].dropna().unique()
    ):

        country_df = df[
            df["country"] == country
        ]

        for method_name, column in methods.items():

            if column not in df.columns:
                continue

            result = evaluate(
                country_df,
                column
            )

            country_rows.append({
                "Country": country,
                "Method": method_name,
                "Records": len(country_df),
                **result
            })

country_results = pd.DataFrame(country_rows)

print("\n==============================================")
print("COMPARISON BY COUNTRY")
print("==============================================")

display(country_results)


# ------------------------------------------------------------
# 7. Role comparison
# ------------------------------------------------------------
role_rows = []

if "title" in df.columns:

    for role in sorted(
        df["title"].dropna().unique()
    ):

        role_df = df[
            df["title"] == role
        ]

        for method_name, column in methods.items():

            if column not in df.columns:
                continue

            result = evaluate(
                role_df,
                column
            )

            role_rows.append({
                "Role": role,
                "Method": method_name,
                "Records": len(role_df),
                **result
            })

role_results = pd.DataFrame(role_rows)

print("\n==============================================")
print("COMPARISON BY ROLE")
print("==============================================")

display(role_results)


# ------------------------------------------------------------
# 8. Zero-skill comparison
# ------------------------------------------------------------
zero_rows = []

for method_name, column in methods.items():

    if column not in df.columns:
        continue

    counts = df[column].apply(
        lambda x: len(parse_skills(x))
    )

    zero_count = int(
        (counts == 0).sum()
    )

    zero_rows.append({
        "Method": method_name,
        "Total Records": len(df),
        "Records With Skills": int(
            (counts > 0).sum()
        ),
        "Zero Skill Records": zero_count,
        "Zero Skill Percentage": round(
            zero_count / len(df) * 100,
            2
        ) if len(df) else 0
    })

zero_results = pd.DataFrame(zero_rows)

print("\n==============================================")
print("ZERO-SKILL COMPARISON")
print("==============================================")

display(zero_results)


# ------------------------------------------------------------
# 9. Create method summary
# ------------------------------------------------------------
method_summary = overall_results[
    [
        "Method",
        "Precision",
        "Recall",
        "F1"
    ]
].copy()

print("\n==============================================")
print("TASK 3 METHOD SUMMARY")
print("==============================================")

display(method_summary)


# ------------------------------------------------------------
# 10. Save everything
# ------------------------------------------------------------
output_file = (
    "26_Task3_Three_Method_Comparison.xlsx"
)

with pd.ExcelWriter(
    output_file,
    engine="openpyxl"
) as writer:

    overall_results.to_excel(
        writer,
        sheet_name="Overall",
        index=False
    )

    country_results.to_excel(
        writer,
        sheet_name="By_Country",
        index=False
    )

    role_results.to_excel(
        writer,
        sheet_name="By_Role",
        index=False
    )

    zero_results.to_excel(
        writer,
        sheet_name="Zero_Skill",
        index=False
    )

    method_summary.to_excel(
        writer,
        sheet_name="Method_Summary",
        index=False
    )

    df.to_excel(
        writer,
        sheet_name="Detailed_Results",
        index=False
    )


# ------------------------------------------------------------
# 11. Download
# ------------------------------------------------------------
from google.colab import files

files.download(output_file)

print("\n✅ CELL 48 COMPLETED")
print("📥 Download:", output_file)
```

```python
# ============================================================
# CELL 50 — SKILL EXTRACTION TRACEABILITY
# ============================================================

import pandas as pd
import re
import ast
from datetime import datetime

# ------------------------------------------------------------
# 1. Select dataframe
# ------------------------------------------------------------
if "transformer_df" in globals():
    df = transformer_df.copy()
elif "semantic_df" in globals():
    df = semantic_df.copy()
else:
    raise ValueError(
        "❌ Required dataframe not found. Run previous cells first."
    )

print("Records:", len(df))


# ------------------------------------------------------------
# 2. Skill parser
# ------------------------------------------------------------
def parse_skills(value):

    if value is None:
        return []

    if isinstance(value, float) and pd.isna(value):
        return []

    if isinstance(value, (list, tuple, set)):
        return [
            str(x).strip()
            for x in value
            if str(x).strip()
        ]

    text = str(value).strip()

    if not text:
        return []

    try:
        parsed = ast.literal_eval(text)

        if isinstance(parsed, (list, tuple, set)):
            return [
                str(x).strip()
                for x in parsed
                if str(x).strip()
            ]

    except:
        pass

    return [
        x.strip()
        for x in text.split(",")
        if x.strip()
    ]


# ------------------------------------------------------------
# 3. Normalization function
# ------------------------------------------------------------
def normalize_skill(skill):

    if not skill:
        return None

    original = str(skill).strip()

    key = original.lower()

    # Cybersecurity normalization
    if "CYBERSECURITY_NORMALIZATION" in globals():

        mapping = CYBERSECURITY_NORMALIZATION

        if key in mapping:
            return mapping[key]

        key_clean = re.sub(
            r"\s+",
            " ",
            key
        )

        if key_clean in mapping:
            return mapping[key_clean]

    return original


# ------------------------------------------------------------
# 4. Find job text
# ------------------------------------------------------------
if "jobDescription" in df.columns:
    text_column = "jobDescription"

elif "job_description" in df.columns:
    text_column = "job_description"

elif "nlp_text" in df.columns:
    text_column = "nlp_text"

else:
    raise ValueError(
        "❌ No job description/text column found."
    )


# ------------------------------------------------------------
# 5. Create traceability records
# ------------------------------------------------------------
traceability_records = []


def find_source_phrase(text, skill):

    if not text or not skill:
        return ""

    text_lower = str(text).lower()
    skill_lower = str(skill).lower()

    position = text_lower.find(skill_lower)

    if position >= 0:
        start = max(0, position - 80)
        end = min(
            len(text),
            position + len(skill) + 80
        )

        return str(text)[start:end]

    # Try normalized spacing
    pattern = re.escape(skill_lower)
    pattern = pattern.replace(r"\ ", r"\s+")

    match = re.search(
        pattern,
        text_lower
    )

    if match:

        start = max(
            0,
            match.start() - 80
        )

        end = min(
            len(text),
            match.end() + 80
        )

        return str(text)[start:end]

    return ""


# ------------------------------------------------------------
# 6. Add traceability for each method
# ------------------------------------------------------------
method_columns = {
    "Baseline Dictionary/Regex": "baseline_skills",
    "Sentence Transformer": "semantic_skills",
    "Transformer NER": "transformer_skills"
}


for _, row in df.iterrows():

    job_id = row.get(
        "jobId",
        ""
    )

    title = row.get(
        "title",
        ""
    )

    country = row.get(
        "country",
        ""
    )

    text = row.get(
        text_column,
        ""
    )

    if pd.isna(text):
        text = ""

    text = str(text)

    for method, skill_column in method_columns.items():

        if skill_column not in df.columns:
            continue

        skills = parse_skills(
            row[skill_column]
        )

        for skill in skills:

            normalized = normalize_skill(
                skill
            )

            source_phrase = find_source_phrase(
                text,
                skill
            )

            # Semantic confidence
            confidence = None

            if (
                method == "Sentence Transformer"
                and "semantic_confidence" in df.columns
            ):

                conf_value = row.get(
                    "semantic_confidence"
                )

                if isinstance(
                    conf_value,
                    (list, tuple)
                ):

                    try:

                        skill_index = skills.index(
                            skill
                        )

                        if skill_index < len(
                            conf_value
                        ):
                            confidence = float(
                                conf_value[
                                    skill_index
                                ]
                            )

                    except:
                        confidence = None

            traceability_records.append({

                "jobId": job_id,

                "title": title,

                "country": country,

                "method": method,

                "original_skill": skill,

                "normalized_skill": normalized,

                "source_text_column": text_column,

                "source_text": source_phrase,

                "confidence_or_similarity": confidence,

                "model": (
                    row.get(
                        "semantic_model",
                        ""
                    )
                    if method ==
                    "Sentence Transformer"
                    else (
                        row.get(
                            "transformer_model",
                            ""
                        )
                        if method ==
                        "Transformer NER"
                        else "Dictionary/Regex"
                    )
                ),

                "processing_date":
                    datetime.now().strftime(
                        "%Y-%m-%d"
                    )
            })


# ------------------------------------------------------------
# 7. Convert to dataframe
# ------------------------------------------------------------
traceability_df = pd.DataFrame(
    traceability_records
)


# ------------------------------------------------------------
# 8. Remove exact duplicate trace records
# ------------------------------------------------------------
if len(traceability_df) > 0:

    traceability_df = (
        traceability_df
        .drop_duplicates(
            subset=[
                "jobId",
                "method",
                "original_skill",
                "normalized_skill"
            ]
        )
        .reset_index(drop=True)
    )


# ------------------------------------------------------------
# 9. Summary
# ------------------------------------------------------------
print("\n==============================================")
print("TRACEABILITY SUMMARY")
print("==============================================")

print(
    "Traceability records:",
    len(traceability_df)
)

if len(traceability_df) > 0:

    print(
        "Unique jobs:",
        traceability_df["jobId"].nunique()
    )

    print(
        "Unique normalized skills:",
        traceability_df[
            "normalized_skill"
        ].nunique()
    )

    print("\nRecords by method:")

    display(
        traceability_df[
            "method"
        ].value_counts()
        .reset_index()
        .rename(
            columns={
                "count": "Records"
            }
        )
    )


# ------------------------------------------------------------
# 10. Save traceability artifact
# ------------------------------------------------------------
output_file = (
    "28_Skill_Extraction_Traceability.xlsx"
)

with pd.ExcelWriter(
    output_file,
    engine="openpyxl"
) as writer:

    traceability_df.to_excel(
        writer,
        sheet_name="Traceability",
        index=False
    )

    if len(traceability_df) > 0:

        traceability_df.groupby(
            "method"
        ).size().reset_index(
            name="Traceability_Records"
        ).to_excel(
            writer,
            sheet_name="Method_Summary",
            index=False
        )


# ------------------------------------------------------------
# 11. Download
# ------------------------------------------------------------
from google.colab import files

files.download(output_file)

print("\n✅ CELL 50 COMPLETED")
print("📥 Download:", output_file)
```

```python
# ============================================================
# CELL 54 — RESUME SKILL EXTRACTION EVALUATION
# ============================================================

import pandas as pd
import ast
from google.colab import files

print("==============================================")
print("RESUME SKILL EXTRACTION EVALUATION")
print("==============================================")

# ------------------------------------------------------------
# FIND VERIFIED RESUME BENCHMARK
# ------------------------------------------------------------

if "resume_benchmark" not in globals():
    raise ValueError(
        "❌ Verified resume benchmark not found.\n"
        "Please run Cell 53 successfully first."
    )

df_resume_eval = resume_benchmark.copy()

# ------------------------------------------------------------
# PARSE SKILLS
# ------------------------------------------------------------

def parse_skills(value):

    if isinstance(value, list):
        return {
            str(x).strip().lower()
            for x in value
            if str(x).strip()
        }

    if pd.isna(value):
        return set()

    text = str(value).strip()

    if not text:
        return set()

    try:
        parsed = ast.literal_eval(text)

        if isinstance(parsed, list):
            return {
                str(x).strip().lower()
                for x in parsed
                if str(x).strip()
            }

    except:
        pass

    text = (
        text.replace("[", "")
            .replace("]", "")
            .replace("'", "")
            .replace('"', "")
    )

    if ";" in text:
        parts = text.split(";")
    elif "|" in text:
        parts = text.split("|")
    elif "," in text:
        parts = text.split(",")
    else:
        parts = [text]

    return {
        x.strip().lower()
        for x in parts
        if x.strip()
    }


# ------------------------------------------------------------
# FIND PREDICTED SKILLS
# ------------------------------------------------------------

possible_prediction_columns = [
    "candidate_skills",
    "extracted_skills",
    "normalized_skills",
    "predicted_skills"
]

prediction_column = None

for col in possible_prediction_columns:
    if col in df_resume_eval.columns:
        prediction_column = col
        break

if prediction_column is None:

    raise ValueError(
        "❌ No predicted skill column found.\n"
        "Expected one of:\n"
        + ", ".join(possible_prediction_columns)
    )

print("Predicted skill column:", prediction_column)

# ------------------------------------------------------------
# CREATE SETS
# ------------------------------------------------------------

df_resume_eval["gold_set"] = (
    df_resume_eval["gold_candidate_skills"]
    .apply(parse_skills)
)

df_resume_eval["predicted_set"] = (
    df_resume_eval[prediction_column]
    .apply(parse_skills)
)

# ------------------------------------------------------------
# CALCULATE METRICS
# ------------------------------------------------------------

results = []

total_tp = 0
total_fp = 0
total_fn = 0

for _, row in df_resume_eval.iterrows():

    gold = row["gold_set"]
    predicted = row["predicted_set"]

    tp = len(gold & predicted)
    fp = len(predicted - gold)
    fn = len(gold - predicted)

    total_tp += tp
    total_fp += fp
    total_fn += fn

    if tp + fp > 0:
        precision = tp / (tp + fp)
    else:
        precision = 0

    if tp + fn > 0:
        recall = tp / (tp + fn)
    else:
        recall = 0

    if precision + recall > 0:
        f1 = (
            2 * precision * recall
            / (precision + recall)
        )
    else:
        f1 = 0

    results.append({
        "candidateId": row["candidateId"],
        "gold_skill_count": len(gold),
        "predicted_skill_count": len(predicted),
        "TP": tp,
        "FP": fp,
        "FN": fn,
        "Precision": round(precision, 4),
        "Recall": round(recall, 4),
        "F1": round(f1, 4)
    })

candidate_results = pd.DataFrame(results)

# ------------------------------------------------------------
# OVERALL METRICS
# ------------------------------------------------------------

if total_tp + total_fp > 0:
    overall_precision = total_tp / (
        total_tp + total_fp
    )
else:
    overall_precision = 0

if total_tp + total_fn > 0:
    overall_recall = total_tp / (
        total_tp + total_fn
    )
else:
    overall_recall = 0

if overall_precision + overall_recall > 0:
    overall_f1 = (
        2 * overall_precision * overall_recall
        / (overall_precision + overall_recall)
    )
else:
    overall_f1 = 0

overall_results = pd.DataFrame([{
    "Dataset": "Resume Benchmark",
    "Candidates": len(df_resume_eval),
    "TP": total_tp,
    "FP": total_fp,
    "FN": total_fn,
    "Precision": round(overall_precision, 4),
    "Recall": round(overall_recall, 4),
    "F1": round(overall_f1, 4)
}])

# ------------------------------------------------------------
# ZERO-SKILL ANALYSIS
# ------------------------------------------------------------

zero_gold = (
    df_resume_eval["gold_set"]
    .apply(len)
    .eq(0)
    .sum()
)

zero_predicted = (
    df_resume_eval["predicted_set"]
    .apply(len)
    .eq(0)
    .sum()
)

zero_skill_summary = pd.DataFrame([{
    "Total_candidates": len(df_resume_eval),
    "Zero_gold_skill_candidates": zero_gold,
    "Zero_predicted_skill_candidates": zero_predicted,
    "Gold_zero_skill_rate": round(
        zero_gold / len(df_resume_eval), 4
    ),
    "Predicted_zero_skill_rate": round(
        zero_predicted / len(df_resume_eval), 4
    )
}])

# ------------------------------------------------------------
# DISPLAY RESULTS
# ------------------------------------------------------------

print("\n==============================================")
print("OVERALL RESUME EXTRACTION RESULTS")
print("==============================================")

print(
    overall_results.to_string(index=False)
)

print("\n==============================================")
print("ZERO-SKILL ANALYSIS")
print("==============================================")

print(
    zero_skill_summary.to_string(index=False)
)

print("\n==============================================")
print("PER-CANDIDATE RESULTS")
print("==============================================")

display(
    candidate_results.head(20)
)

# ------------------------------------------------------------
# SAVE RESULTS
# ------------------------------------------------------------

output_file = "32_Resume_Skill_Extraction_Evaluation.xlsx"

with pd.ExcelWriter(output_file, engine="openpyxl") as writer:

    overall_results.to_excel(
        writer,
        sheet_name="Overall_Metrics",
        index=False
    )

    candidate_results.to_excel(
        writer,
        sheet_name="Candidate_Metrics",
        index=False
    )

    zero_skill_summary.to_excel(
        writer,
        sheet_name="Zero_Skill_Analysis",
        index=False
    )

    df_resume_eval.drop(
        columns=["gold_set", "predicted_set"],
        errors="ignore"
    ).to_excel(
        writer,
        sheet_name="Evaluation_Data",
        index=False
    )

print("\n💾 Saved:", output_file)

# ------------------------------------------------------------
# DOWNLOAD
# ------------------------------------------------------------

files.download(output_file)

print("\n⬇️ Download started.")
print("✅ Resume skill extraction evaluation completed.")
```

```python
# ============================================================
# CELL 55 — SEMANTIC RESUME SKILL EXTRACTION
# ============================================================

!pip -q install sentence-transformers

import pandas as pd
import numpy as np
from sentence_transformers import SentenceTransformer, util
from google.colab import files

print("==============================================")
print("SEMANTIC RESUME SKILL EXTRACTION")
print("==============================================")

# ------------------------------------------------------------
# 1. CHECK VERIFIED BENCHMARK
# ------------------------------------------------------------

if "resume_benchmark" not in globals():
    raise ValueError(
        "❌ Verified resume benchmark not found.\n"
        "Run Cell 53 successfully first."
    )

resume_semantic_df = resume_benchmark.copy()

# ------------------------------------------------------------
# 2. BUILD SKILL VOCABULARY
# ------------------------------------------------------------

skill_values = []

possible_skill_sources = [
    "gold_candidate_skills",
    "candidate_skills"
]

# Resume skills
for col in possible_skill_sources:
    if col in resume_semantic_df.columns:
        for value in resume_semantic_df[col].dropna():

            if isinstance(value, list):
                skill_values.extend(value)

            else:
                text = str(value)

                for separator in [",", ";", "|"]:
                    if separator in text:
                        skill_values.extend(
                            text.split(separator)
                        )
                        break
                else:
                    skill_values.append(text)

# Job dataset skills if available
job_dataframes = [
    "df",
    "evaluation_df",
    "semantic_df",
    "transformer_df",
    "final_unified_dataset"
]

for df_name in job_dataframes:

    if df_name in globals():

        source_df = globals()[df_name]

        for col in [
            "normalized_skills",
            "extracted_skills",
            "tagsAndSkills"
        ]:

            if col in source_df.columns:

                for value in source_df[col].dropna():

                    if isinstance(value, list):
                        skill_values.extend(value)

                    else:
                        text = str(value)

                        for separator in [",", ";", "|"]:
                            if separator in text:
                                skill_values.extend(
                                    text.split(separator)
                                )
                                break
                        else:
                            skill_values.append(text)

# ------------------------------------------------------------
# 3. CYBERSECURITY VOCABULARY
# ------------------------------------------------------------

cybersecurity_skills = [
    "Cybersecurity",
    "Information Security",
    "Network Security",
    "Application Security",
    "Cloud Security",
    "Security Operations",
    "SOC",
    "SIEM",
    "IAM",
    "PAM",
    "Vulnerability Assessment",
    "Penetration Testing",
    "Incident Response",
    "Digital Forensics",
    "Threat Intelligence",
    "Security Architecture",
    "Zero Trust",
    "Endpoint Security",
    "EDR",
    "XDR",
    "IDS",
    "IPS",
    "Firewall",
    "Cryptography",
    "Encryption",
    "ISO 27001",
    "NIST",
    "OWASP"
]

skill_values.extend(cybersecurity_skills)

# ------------------------------------------------------------
# 4. CLEAN VOCABULARY
# ------------------------------------------------------------

skill_vocabulary = sorted({
    str(skill).strip()
    for skill in skill_values
    if str(skill).strip()
    and len(str(skill).strip()) >= 2
})

print("Skill vocabulary size:", len(skill_vocabulary))

if len(skill_vocabulary) == 0:
    raise ValueError(
        "❌ No skill vocabulary available."
    )

# ------------------------------------------------------------
# 5. LOAD SENTENCE TRANSFORMER
# ------------------------------------------------------------

print("\nLoading semantic model...")

semantic_model = SentenceTransformer(
    "all-MiniLM-L6-v2"
)

print("✅ Semantic model loaded.")

# ------------------------------------------------------------
# 6. CREATE SKILL EMBEDDINGS
# ------------------------------------------------------------

print("\nCreating skill embeddings...")

skill_embeddings = semantic_model.encode(
    skill_vocabulary,
    convert_to_tensor=True,
    show_progress_bar=True
)

# ------------------------------------------------------------
# 7. ENCODE RESUMES
# ------------------------------------------------------------

resume_texts = (
    resume_semantic_df["resume_text"]
    .fillna("")
    .astype(str)
    .tolist()
)

print("\nEncoding candidate resumes...")

resume_embeddings = semantic_model.encode(
    resume_texts,
    convert_to_tensor=True,
    batch_size=16,
    show_progress_bar=True
)

# ------------------------------------------------------------
# 8. SEMANTIC SKILL MATCHING
# ------------------------------------------------------------

SEMANTIC_THRESHOLD = 0.55
TOP_K = 20

semantic_skill_results = []
semantic_confidence_results = []

for i, resume_embedding in enumerate(resume_embeddings):

    similarities = util.cos_sim(
        resume_embedding,
        skill_embeddings
    )[0]

    top_k = min(
        TOP_K,
        len(skill_vocabulary)
    )

    top_values, top_indices = similarities.topk(top_k)

    selected_skills = []
    selected_scores = []

    for score, index in zip(
        top_values.cpu().tolist(),
        top_indices.cpu().tolist()
    ):

        if score >= SEMANTIC_THRESHOLD:

            selected_skills.append(
                skill_vocabulary[index]
            )

            selected_scores.append(
                round(float(score), 4)
            )

    semantic_skill_results.append(
        selected_skills
    )

    semantic_confidence_results.append(
        selected_scores
    )

# ------------------------------------------------------------
# 9. STORE RESULTS
# ------------------------------------------------------------

resume_semantic_df["semantic_candidate_skills"] = (
    semantic_skill_results
)

resume_semantic_df["semantic_confidence"] = (
    semantic_confidence_results
)

resume_semantic_df["semantic_confidence_score"] = (
    resume_semantic_df["semantic_confidence"]
    .apply(
        lambda x: max(x) if len(x) > 0 else 0
    )
)

resume_semantic_df["semantic_model"] = (
    "sentence-transformers/all-MiniLM-L6-v2"
)

resume_semantic_df["semantic_threshold"] = (
    SEMANTIC_THRESHOLD
)

resume_semantic_df["semantic_method"] = (
    "Sentence Transformer cosine similarity"
)

# ------------------------------------------------------------
# 10. ZERO-SKILL ANALYSIS
# ------------------------------------------------------------

zero_semantic = (
    resume_semantic_df[
        "semantic_candidate_skills"
    ]
    .apply(len)
    .eq(0)
    .sum()
)

print("\n==============================================")
print("SEMANTIC EXTRACTION SUMMARY")
print("==============================================")

print(
    "Total candidates:",
    len(resume_semantic_df)
)

print(
    "Candidates with semantic skills:",
    len(resume_semantic_df) - zero_semantic
)

print(
    "Candidates with zero semantic skills:",
    zero_semantic
)

print(
    "Zero-skill rate:",
    round(
        zero_semantic / len(resume_semantic_df),
        4
    )
)

# ------------------------------------------------------------
# 11. DISPLAY SAMPLE
# ------------------------------------------------------------

display(
    resume_semantic_df[
        [
            "candidateId",
            "semantic_candidate_skills",
            "semantic_confidence_score"
        ]
    ].head(10)
)

# ------------------------------------------------------------
# 12. SAVE OUTPUT
# ------------------------------------------------------------

output_file = (
    "33_Semantic_Resume_Skill_Extraction.xlsx"
)

resume_semantic_df.to_excel(
    output_file,
    index=False
)

print("\n💾 Saved:", output_file)

# ------------------------------------------------------------
# 13. DOWNLOAD
# ------------------------------------------------------------

files.download(output_file)

print("\n⬇️ Download started.")
print("✅ Semantic resume skill extraction completed.")
```

```python
# ============================================================
# CELL 58 — WEIGHTED SKILL-GAP ANALYSIS
# ============================================================

import pandas as pd
import numpy as np
import ast
from datetime import datetime
from google.colab import files

print("==============================================")
print("WEIGHTED SKILL-GAP ANALYSIS")
print("==============================================")

# ------------------------------------------------------------
# 1. CHECK REQUIRED DATA
# ------------------------------------------------------------

if "resume_semantic_df" not in globals():
    raise ValueError(
        "❌ Semantic resume data not found.\n"
        "Run Cell 55 first."
    )

if "resume_benchmark" not in globals():
    raise ValueError(
        "❌ Resume benchmark not found.\n"
        "Run Cell 53 first."
    )

resume_gap_df = resume_semantic_df.copy()

# ------------------------------------------------------------
# 2. SKILL PARSER
# ------------------------------------------------------------

def parse_skills(value):

    if isinstance(value, list):
        return {
            str(x).strip().lower()
            for x in value
            if str(x).strip()
        }

    if pd.isna(value):
        return set()

    text = str(value).strip()

    if not text:
        return set()

    try:
        parsed = ast.literal_eval(text)

        if isinstance(parsed, list):
            return {
                str(x).strip().lower()
                for x in parsed
                if str(x).strip()
            }

    except:
        pass

    text = (
        text.replace("[", "")
        .replace("]", "")
        .replace("'", "")
        .replace('"', "")
    )

    if ";" in text:
        parts = text.split(";")
    elif "|" in text:
        parts = text.split("|")
    elif "," in text:
        parts = text.split(",")
    else:
        parts = [text]

    return {
        x.strip().lower()
        for x in parts
        if x.strip()
    }

# ------------------------------------------------------------
# 3. GOLD AND CANDIDATE SKILLS
# ------------------------------------------------------------

resume_gap_df["required_skills"] = (
    resume_gap_df["gold_candidate_skills"]
    .apply(parse_skills)
)

resume_gap_df["candidate_skills"] = (
    resume_gap_df["semantic_candidate_skills"]
    .apply(parse_skills)
)

# ------------------------------------------------------------
# 4. BUILD MARKET FREQUENCY
# ------------------------------------------------------------

market_frequency = {}

# Use the complete job dataset if available
job_sources = [
    "df",
    "evaluation_df",
    "semantic_df",
    "transformer_df",
    "final_unified_dataset"
]

job_source = None

for source_name in job_sources:

    if source_name in globals():

        possible_df = globals()[source_name]

        if isinstance(possible_df, pd.DataFrame):
            job_source = possible_df
            break

if job_source is not None:

    print(
        "Using job-market data:",
        len(job_source),
        "records"
    )

    skill_columns = [
        "normalized_skills",
        "extracted_skills",
        "tagsAndSkills"
    ]

    selected_skill_column = None

    for col in skill_columns:

        if col in job_source.columns:
            selected_skill_column = col
            break

    if selected_skill_column:

        for value in job_source[
            selected_skill_column
        ].dropna():

            skills = parse_skills(value)

            for skill in skills:
                market_frequency[skill] = (
                    market_frequency.get(skill, 0) + 1
                )

else:

    print(
        "⚠️ Complete job dataset not found."
    )

    # Fallback to benchmark frequency
    for skills in resume_gap_df["required_skills"]:

        for skill in skills:

            market_frequency[skill] = (
                market_frequency.get(skill, 0) + 1
            )

# ------------------------------------------------------------
# 5. NORMALIZE MARKET FREQUENCY
# ------------------------------------------------------------

if market_frequency:

    max_frequency = max(
        market_frequency.values()
    )

else:
    max_frequency = 1

# ------------------------------------------------------------
# 6. ROLE IMPORTANCE
# ------------------------------------------------------------
#
# Role importance is based on how frequently a required
# skill appears across the validated resume benchmark.
#
# This avoids assigning the same importance to every skill.
# ------------------------------------------------------------

skill_required_count = {}

for skills in resume_gap_df["required_skills"]:

    for skill in skills:

        skill_required_count[skill] = (
            skill_required_count.get(skill, 0) + 1
        )

max_required = max(
    skill_required_count.values()
) if skill_required_count else 1

# ------------------------------------------------------------
# 7. RECENCY WEIGHT
# ------------------------------------------------------------
#
# If job posting dates are available, newer postings receive
# greater importance.
#
# Otherwise a neutral recency weight of 1.0 is used.
# ------------------------------------------------------------

def calculate_recency_weight():

    if job_source is None:
        return 1.0

    date_columns = [
        "datePosted",
        "postedDate",
        "date_posted",
        "processing_date"
    ]

    date_column = None

    for col in date_columns:

        if col in job_source.columns:
            date_column = col
            break

    if date_column is None:
        return 1.0

    dates = pd.to_datetime(
        job_source[date_column],
        errors="coerce"
    )

    valid_dates = dates.dropna()

    if len(valid_dates) == 0:
        return 1.0

    latest_date = valid_dates.max()

    age_days = (
        datetime.now() - latest_date
    ).days

    # Exponential-style recency decay
    recency = np.exp(
        -age_days / 365
    )

    return float(
        max(0.25, min(1.0, recency))
    )

global_recency_weight = calculate_recency_weight()

# ------------------------------------------------------------
# 8. CANDIDATE PROFICIENCY
# ------------------------------------------------------------
#
# Since the current resume dataset does not necessarily have
# validated proficiency levels, we infer a conservative
# evidence-based proficiency score:
#
# 1.00 = explicitly/strongly matched
# 0.75 = moderate semantic match
# 0.50 = weak semantic evidence
# 0.25 = no evidence
#
# This is NOT a certified proficiency assessment.
# ------------------------------------------------------------

def proficiency_score(skill, row):

    semantic_skills = parse_skills(
        row.get("semantic_candidate_skills", [])
    )

    if skill in semantic_skills:

        confidence_values = row.get(
            "semantic_confidence",
            []
        )

        if isinstance(confidence_values, list):

            # Find matching skill position where possible
            try:

                skill_index = [
                    str(x).strip().lower()
                    for x in row[
                        "semantic_candidate_skills"
                    ]
                ].index(skill)

                if skill_index < len(confidence_values):

                    similarity = float(
                        confidence_values[
                            skill_index
                        ]
                    )

                    if similarity >= 0.75:
                        return 1.00

                    elif similarity >= 0.65:
                        return 0.75

                    else:
                        return 0.50

            except:
                pass

        return 0.75

    return 0.25

# ------------------------------------------------------------
# 9. WEIGHTED GAP FORMULA
# ------------------------------------------------------------
#
# Skill Weight =
#   0.35 Market Frequency
# + 0.30 Role Importance
# + 0.20 Recency
# + 0.15 Skill Importance Baseline
#
# Gap Contribution =
#   Skill Weight × (1 - Candidate Proficiency)
#
# Weighted Gap Score =
#   Sum(Gap Contribution) /
#   Sum(Skill Weight)
#
# Therefore:
#   0.00 = no weighted gap
#   1.00 = maximum weighted gap
# ------------------------------------------------------------

gap_rows = []

for _, row in resume_gap_df.iterrows():

    candidate_id = row["candidateId"]

    required = row["required_skills"]
    candidate = row["candidate_skills"]

    missing_skills = required - candidate

    skill_details = []

    weighted_gap_total = 0
    total_required_weight = 0

    for skill in required:

        # Market frequency
        market_score = (
            market_frequency.get(skill, 0)
            / max_frequency
        )

        # Role importance
        role_score = (
            skill_required_count.get(skill, 0)
            / max_required
        )

        # Recency
        recency_score = global_recency_weight

        # Baseline importance
        importance_score = 1.0

        # Combined weight
        skill_weight = (
            0.35 * market_score
            + 0.30 * role_score
            + 0.20 * recency_score
            + 0.15 * importance_score
        )

        proficiency = proficiency_score(
            skill,
            row
        )

        gap_contribution = (
            skill_weight *
            (1 - proficiency)
        )

        total_required_weight += skill_weight
        weighted_gap_total += gap_contribution

        skill_details.append({
            "skill": skill,
            "market_frequency":
                market_frequency.get(skill, 0),
            "market_score":
                round(market_score, 4),
            "role_importance":
                round(role_score, 4),
            "recency_weight":
                round(recency_score, 4),
            "candidate_proficiency":
                round(proficiency, 4),
            "skill_weight":
                round(skill_weight, 4),
            "gap_contribution":
                round(gap_contribution, 4),
            "missing":
                skill in missing_skills
        })

    if total_required_weight > 0:

        weighted_gap_score = (
            weighted_gap_total /
            total_required_weight
        )

    else:
        weighted_gap_score = 0

    gap_rows.append({

        "candidateId":
            candidate_id,

        "required_skill_count":
            len(required),

        "candidate_skill_count":
            len(candidate),

        "missing_skill_count":
            len(missing_skills),

        "missing_skills":
            sorted(missing_skills),

        "weighted_gap_score":
            round(
                weighted_gap_score,
                4
            ),

        "weighted_gap_percentage":
            round(
                weighted_gap_score * 100,
                2
            ),

        "skill_weight_details":
            skill_details
    })

# ------------------------------------------------------------
# 10. CREATE OUTPUT DATAFRAME
# ------------------------------------------------------------

weighted_gap_df = pd.DataFrame(
    gap_rows
)

# ------------------------------------------------------------
# 11. DISPLAY RESULTS
# ------------------------------------------------------------

print("\n==============================================")
print("WEIGHTED SKILL-GAP RESULTS")
print("==============================================")

display(
    weighted_gap_df[
        [
            "candidateId",
            "required_skill_count",
            "candidate_skill_count",
            "missing_skill_count",
            "missing_skills",
            "weighted_gap_score",
            "weighted_gap_percentage"
        ]
    ].head(20)
)

# ------------------------------------------------------------
# 12. SAVE FORMULA DOCUMENTATION
# ------------------------------------------------------------

formula_documentation = pd.DataFrame([

    {
        "Component":
            "Market Frequency",
        "Weight":
            0.35,
        "Definition":
            "Relative frequency of a skill in the available job-market dataset."
    },

    {
        "Component":
            "Role Importance",
        "Weight":
            0.30,
        "Definition":
            "Relative frequency with which the skill is required in the validated benchmark."
    },

    {
        "Component":
            "Recency",
        "Weight":
            0.20,
        "Definition":
            "Newer job-market information receives greater weight when posting dates are available."
    },

    {
        "Component":
            "Baseline Importance",
        "Weight":
            0.15,
        "Definition":
            "Neutral baseline importance used when no explicit role-specific importance annotation exists."
    },

    {
        "Component":
            "Candidate Proficiency",
        "Weight":
            "Applied to gap",
        "Definition":
            "Evidence-based estimate derived from semantic similarity; not a certified proficiency assessment."
    },

    {
        "Component":
            "Weighted Gap Score",
        "Weight":
            "Final",
        "Definition":
            "Sum of weighted skill gaps divided by total required skill weight."
    }
])

# ------------------------------------------------------------
# 13. SAVE OUTPUT
# ------------------------------------------------------------

output_file = (
    "36_Weighted_Skill_Gap_Analysis.xlsx"
)

with pd.ExcelWriter(
    output_file,
    engine="openpyxl"
) as writer:

    weighted_gap_df[
        [
            "candidateId",
            "required_skill_count",
            "candidate_skill_count",
            "missing_skill_count",
            "missing_skills",
            "weighted_gap_score",
            "weighted_gap_percentage"
        ]
    ].to_excel(
        writer,
        sheet_name="Weighted_Gap",
        index=False
    )

    formula_documentation.to_excel(
        writer,
        sheet_name="Formula_Documentation",
        index=False
    )

print("\n💾 Saved:", output_file)

# ------------------------------------------------------------
# 14. DOWNLOAD
# ------------------------------------------------------------

files.download(output_file)

print("\n⬇️ Download started.")
print("✅ Weighted skill-gap analysis completed.")
```

```python
# ============================================================
# CELL 59 — RESUME SKILL-GAP TRACEABILITY
# ============================================================

import pandas as pd
import ast
import re
from datetime import datetime
from google.colab import files

print("==============================================")
print("RESUME SKILL-GAP TRACEABILITY")
print("==============================================")

# ------------------------------------------------------------
# 1. CHECK REQUIRED DATA
# ------------------------------------------------------------

if "resume_semantic_df" not in globals():
    raise ValueError(
        "❌ Semantic resume data not found.\n"
        "Run Cell 55 first."
    )

if "weighted_gap_df" not in globals():
    raise ValueError(
        "❌ Weighted skill-gap results not found.\n"
        "Run Cell 58 first."
    )

trace_df = resume_semantic_df.copy()

# ------------------------------------------------------------
# 2. CYBERSECURITY NORMALIZATION
# ------------------------------------------------------------

CYBERSECURITY_NORMALIZATION = {

    "cyber security": "Cybersecurity",
    "cybersecurity": "Cybersecurity",

    "information security": "Information Security",
    "infosec": "Information Security",

    "network security": "Network Security",

    "application security": "Application Security",
    "appsec": "Application Security",

    "cloud security": "Cloud Security",

    "security operations": "Security Operations",
    "secops": "Security Operations",

    "security operations center":
        "Security Operations",

    "soc": "Security Operations",

    "siem": "SIEM",

    "iam": "Identity and Access Management",
    "identity and access management":
        "Identity and Access Management",

    "pam": "Privileged Access Management",
    "privileged access management":
        "Privileged Access Management",

    "vulnerability assessment":
        "Vulnerability Assessment",

    "penetration testing":
        "Penetration Testing",

    "penetration test":
        "Penetration Testing",

    "pentesting":
        "Penetration Testing",

    "pentest":
        "Penetration Testing",

    "incident response":
        "Incident Response",

    "digital forensics":
        "Digital Forensics",

    "threat intelligence":
        "Threat Intelligence",

    "security architecture":
        "Security Architecture",

    "zero trust":
        "Zero Trust",

    "endpoint security":
        "Endpoint Security",

    "edr":
        "Endpoint Detection and Response",

    "xdr":
        "Extended Detection and Response",

    "ids":
        "Intrusion Detection System",

    "ips":
        "Intrusion Prevention System",

    "firewall":
        "Firewall",

    "ngfw":
        "Next Generation Firewall",

    "cryptography":
        "Cryptography",

    "encryption":
        "Encryption",

    "iso 27001":
        "ISO 27001",

    "nist":
        "NIST Cybersecurity Framework",

    "nist csf":
        "NIST Cybersecurity Framework",

    "owasp":
        "OWASP"
}

def normalize_skill(skill):

    original = str(skill).strip()

    if not original:
        return ""

    key = original.lower().strip()

    if key in CYBERSECURITY_NORMALIZATION:
        return CYBERSECURITY_NORMALIZATION[key]

    return original

# ------------------------------------------------------------
# 3. PARSE SKILLS
# ------------------------------------------------------------

def parse_skills(value):

    if isinstance(value, list):
        return [
            str(x).strip()
            for x in value
            if str(x).strip()
        ]

    if pd.isna(value):
        return []

    text = str(value).strip()

    if not text:
        return []

    try:

        parsed = ast.literal_eval(text)

        if isinstance(parsed, list):
            return [
                str(x).strip()
                for x in parsed
                if str(x).strip()
            ]

    except:
        pass

    text = (
        text.replace("[", "")
        .replace("]", "")
        .replace("'", "")
        .replace('"', "")
    )

    if ";" in text:
        parts = text.split(";")

    elif "|" in text:
        parts = text.split("|")

    elif "," in text:
        parts = text.split(",")

    else:
        parts = [text]

    return [
        x.strip()
        for x in parts
        if x.strip()
    ]

# ------------------------------------------------------------
# 4. CREATE TRACEABILITY RECORDS
# ------------------------------------------------------------

trace_records = []

for _, row in trace_df.iterrows():

    candidate_id = row["candidateId"]

    resume_text = str(
        row["resume_text"]
    )

    semantic_skills = parse_skills(
        row.get(
            "semantic_candidate_skills",
            []
        )
    )

    confidence_values = row.get(
        "semantic_confidence",
        []
    )

    if not isinstance(confidence_values, list):
        confidence_values = []

    for index, skill in enumerate(
        semantic_skills
    ):

        normalized_skill = normalize_skill(
            skill
        )

        confidence = None

        if index < len(confidence_values):

            try:
                confidence = float(
                    confidence_values[index]
                )
            except:
                confidence = None

        # ----------------------------------------------------
        # Find literal source phrase in resume
        # ----------------------------------------------------

        source_phrase = skill
        source_start = None
        source_end = None

        pattern = re.escape(skill)

        match = re.search(
            pattern,
            resume_text,
            flags=re.IGNORECASE
        )

        if match:

            source_start = match.start()
            source_end = match.end()

            source_phrase = (
                resume_text[
                    source_start:source_end
                ]
            )

            context_start = max(
                0,
                source_start - 80
            )

            context_end = min(
                len(resume_text),
                source_end + 80
            )

            source_context = (
                resume_text[
                    context_start:context_end
                ]
            )

        else:

            source_context = (
                "Semantic match; exact phrase "
                "not found literally in resume text."
            )

        # ----------------------------------------------------
        # Determine gap status
        # ----------------------------------------------------

        gold_skills = parse_skills(
            row.get(
                "gold_candidate_skills",
                []
            )
        )

        normalized_gold = {
            normalize_skill(x).lower()
            for x in gold_skills
        }

        if normalized_skill.lower() in normalized_gold:
            match_status = "MATCHED_GOLD_SKILL"
        else:
            match_status = "SEMANTIC_PREDICTION_ONLY"

        trace_records.append({

            "candidateId":
                candidate_id,

            "method":
                "Sentence Transformer",

            "model":
                row.get(
                    "semantic_model",
                    "all-MiniLM-L6-v2"
                ),

            "original_skill":
                skill,

            "normalized_skill":
                normalized_skill,

            "match_status":
                match_status,

            "confidence_similarity":
                confidence,

            "source_text_column":
                "resume_text",

            "source_phrase":
                source_phrase,

            "source_context":
                source_context,

            "source_start_character":
                source_start,

            "source_end_character":
                source_end,

            "processing_date":
                datetime.now().strftime(
                    "%Y-%m-%d"
                ),

            "processing_version":
                "SkillBridge_Task3_Resume_v1"
        })

# ------------------------------------------------------------
# 5. CREATE TRACEABILITY DATAFRAME
# ------------------------------------------------------------

resume_traceability_df = pd.DataFrame(
    trace_records
)

# Remove duplicates
resume_traceability_df = (
    resume_traceability_df
    .drop_duplicates(
        subset=[
            "candidateId",
            "original_skill",
            "normalized_skill",
            "method"
        ]
    )
    .reset_index(drop=True)
)

# ------------------------------------------------------------
# 6. ADD WEIGHTED GAP INFORMATION
# ------------------------------------------------------------

if len(resume_traceability_df) > 0:

    gap_lookup = weighted_gap_df.set_index(
        "candidateId"
    )[
        [
            "weighted_gap_score",
            "weighted_gap_percentage",
            "missing_skills"
        ]
    ].to_dict("index")

    resume_traceability_df[
        "weighted_gap_score"
    ] = resume_traceability_df[
        "candidateId"
    ].map(
        lambda x:
        gap_lookup.get(
            x,
            {}
        ).get(
            "weighted_gap_score",
            None
        )
    )

    resume_traceability_df[
        "weighted_gap_percentage"
    ] = resume_traceability_df[
        "candidateId"
    ].map(
        lambda x:
        gap_lookup.get(
            x,
            {}
        ).get(
            "weighted_gap_percentage",
            None
        )
    )

# ------------------------------------------------------------
# 7. DISPLAY
# ------------------------------------------------------------

print("\n==============================================")
print("TRACEABILITY SUMMARY")
print("==============================================")

print(
    "Candidates:",
    trace_df["candidateId"].nunique()
)

print(
    "Trace records:",
    len(resume_traceability_df)
)

print(
    "Exact source matches:",
    (
        resume_traceability_df[
            "source_start_character"
        ]
        .notna()
        .sum()
    )
)

print(
    "Semantic-only matches:",
    (
        resume_traceability_df[
            "source_start_character"
        ]
        .isna()
        .sum()
    )
)

display(
    resume_traceability_df.head(20)
)

# ------------------------------------------------------------
# 8. SAVE OUTPUT
# ------------------------------------------------------------

output_file = (
    "37_Resume_Skill_Gap_Traceability.xlsx"
)

with pd.ExcelWriter(
    output_file,
    engine="openpyxl"
) as writer:

    resume_traceability_df.to_excel(
        writer,
        sheet_name="Skill_Traceability",
        index=False
    )

    weighted_gap_df[
        [
            "candidateId",
            "required_skill_count",
            "candidate_skill_count",
            "missing_skill_count",
            "missing_skills",
            "weighted_gap_score",
            "weighted_gap_percentage"
        ]
    ].to_excel(
        writer,
        sheet_name="Weighted_Gap",
        index=False
    )

print("\n💾 Saved:", output_file)

# ------------------------------------------------------------
# 9. DOWNLOAD
# ------------------------------------------------------------

files.download(output_file)

print("\n⬇️ Download started.")
print("✅ Resume traceability completed.")
```

```python
# ============================================================
# CELL 61 — TRANSFORMER NER RESUME SKILL EXTRACTION
# ============================================================

!pip -q install transformers torch

import pandas as pd
import re
from transformers import pipeline
from datetime import datetime
from google.colab import files

print("==============================================")
print("TRANSFORMER RESUME SKILL EXTRACTION")
print("==============================================")

# ------------------------------------------------------------
# 1. FIND RESUME DATA
# ------------------------------------------------------------

if "resume_semantic_df" in globals():
    transformer_resume_df = resume_semantic_df.copy()

elif "resume_dataframe" in globals():
    transformer_resume_df = resume_dataframe.copy()

elif "candidate_df" in globals():
    transformer_resume_df = candidate_df.copy()

else:
    raise ValueError(
        "❌ Resume dataset not found. Run the previous resume cells first."
    )

# ------------------------------------------------------------
# 2. CHECK REQUIRED COLUMNS
# ------------------------------------------------------------

if "candidateId" not in transformer_resume_df.columns:
    raise ValueError(
        "❌ candidateId column not found."
    )

# Detect resume text column
text_candidates = [
    "candidate_profile_text",
    "resume_text",
    "profile_text",
    "text"
]

resume_text_column = None

for col in text_candidates:
    if col in transformer_resume_df.columns:
        resume_text_column = col
        break

if resume_text_column is None:

    object_columns = transformer_resume_df.select_dtypes(
        include=["object"]
    ).columns.tolist()

    if object_columns:

        transformer_resume_df["candidate_profile_text"] = (
            transformer_resume_df[object_columns]
            .fillna("")
            .astype(str)
            .agg(" ".join, axis=1)
        )

        resume_text_column = "candidate_profile_text"

    else:
        raise ValueError(
            "❌ No resume text column found."
        )

print(
    "Resume text column:",
    resume_text_column
)

# ------------------------------------------------------------
# 3. BUILD SKILL VOCABULARY
# ------------------------------------------------------------

skill_vocabulary = set()

source_dfs = [
    transformer_resume_df
]

for name in [
    "df",
    "evaluation_df",
    "semantic_df",
    "transformer_df",
    "final_unified_dataset"
]:

    if name in globals():

        obj = globals()[name]

        if isinstance(obj, pd.DataFrame):
            source_dfs.append(obj)

skill_columns = [
    "candidate_skills",
    "candidate_semantic_skills",
    "validated_candidate_skills",
    "normalized_skills",
    "extracted_skills",
    "semantic_skills",
    "transformer_skills",
    "tagsAndSkills"
]

for source_df in source_dfs:

    for col in skill_columns:

        if col not in source_df.columns:
            continue

        for value in source_df[col].dropna():

            if isinstance(value, list):
                items = value

            else:
                text = str(value)

                try:
                    parsed = eval(text)

                    if isinstance(parsed, list):
                        items = parsed
                    else:
                        items = re.split(
                            r"[,;|]",
                            text
                        )

                except:
                    items = re.split(
                        r"[,;|]",
                        text
                    )

            for skill in items:

                skill = str(skill).strip()

                if 1 < len(skill) <= 80:
                    skill_vocabulary.add(skill)

# ------------------------------------------------------------
# 4. CYBERSECURITY VOCABULARY
# ------------------------------------------------------------

cybersecurity_terms = {

    "cybersecurity",
    "cyber security",
    "information security",
    "infosec",
    "network security",
    "application security",
    "appsec",
    "cloud security",
    "security operations",
    "secops",
    "soc",
    "siem",
    "iam",
    "pam",
    "penetration testing",
    "pentesting",
    "vulnerability assessment",
    "incident response",
    "digital forensics",
    "threat intelligence",
    "security architecture",
    "zero trust",
    "endpoint security",
    "edr",
    "xdr",
    "ids",
    "ips",
    "firewall",
    "ngfw",
    "cryptography",
    "encryption",
    "owasp",
    "nist",
    "iso 27001"
}

skill_vocabulary.update(
    cybersecurity_terms
)

skill_vocabulary = sorted(
    skill_vocabulary,
    key=lambda x: len(x),
    reverse=True
)

print(
    "Skill vocabulary size:",
    len(skill_vocabulary)
)

# ------------------------------------------------------------
# 5. LOAD TRANSFORMER NER MODEL
# ------------------------------------------------------------

print("\nLoading transformer model...")

ner_model = pipeline(
    "ner",
    model="dslim/bert-base-NER",
    aggregation_strategy="simple"
)

print("✅ Transformer model loaded.")

# ------------------------------------------------------------
# 6. NORMALIZATION FUNCTION
# ------------------------------------------------------------

def normalize_skill(skill):

    s = str(skill).lower().strip()

    cybersecurity_map = {

        "cyber security":
            "Cybersecurity",

        "cybersecurity":
            "Cybersecurity",

        "infosec":
            "Information Security",

        "information security":
            "Information Security",

        "appsec":
            "Application Security",

        "application security":
            "Application Security",

        "secops":
            "Security Operations",

        "soc":
            "Security Operations",

        "pentesting":
            "Penetration Testing",

        "pentest":
            "Penetration Testing",

        "vulnerability assessment":
            "Vulnerability Assessment",

        "incident response":
            "Incident Response",

        "threat intelligence":
            "Threat Intelligence",

        "digital forensics":
            "Digital Forensics",

        "zero trust":
            "Zero Trust",

        "endpoint security":
            "Endpoint Security",

        "security architecture":
            "Security Architecture",

        "network security":
            "Network Security",

        "cloud security":
            "Cloud Security"
    }

    return cybersecurity_map.get(
        s,
        skill.strip()
    )

# ------------------------------------------------------------
# 7. MAP NER ENTITIES TO SKILLS
# ------------------------------------------------------------

def extract_transformer_skills(text):

    text = str(text)

    if not text.strip():
        return [], []

    try:

        entities = ner_model(
            text[:5000]
        )

    except Exception:

        return [], []

    detected_entities = []

    transformer_skills = set()

    for entity in entities:

        phrase = entity.get(
            "word",
            ""
        ).strip()

        if not phrase:
            continue

        detected_entities.append(
            phrase
        )

        phrase_norm = normalize_skill(
            phrase
        )

        # Direct vocabulary match
        for skill in skill_vocabulary:

            skill_norm = normalize_skill(
                skill
            )

            if (
                skill_norm.lower()
                == phrase_norm.lower()
            ):

                transformer_skills.add(
                    skill_norm
                )

            elif (
                skill.lower()
                in phrase.lower()
                or
                phrase.lower()
                in skill.lower()
            ):

                transformer_skills.add(
                    skill_norm
                )

    return (
        sorted(set(detected_entities)),
        sorted(transformer_skills)
    )

# ------------------------------------------------------------
# 8. PROCESS RESUMES
# ------------------------------------------------------------

transformer_entities_list = []
transformer_skills_list = []

print("\nProcessing resumes...")

for i, text in enumerate(
    transformer_resume_df[
        resume_text_column
    ].fillna("")
):

    entities, skills = (
        extract_transformer_skills(text)
    )

    transformer_entities_list.append(
        entities
    )

    transformer_skills_list.append(
        skills
    )

    if (i + 1) % 10 == 0:
        print(
            f"Processed {i + 1} resumes"
        )

# ------------------------------------------------------------
# 9. ADD RESULTS
# ------------------------------------------------------------

transformer_resume_df[
    "transformer_entities"
] = transformer_entities_list

transformer_resume_df[
    "transformer_candidate_skills"
] = transformer_skills_list

transformer_resume_df[
    "transformer_model"
] = "dslim/bert-base-NER"

transformer_resume_df[
    "transformer_method"
] = (
    "Transformer NER + Skill Vocabulary Mapping"
)

transformer_resume_df[
    "processing_date"
] = datetime.now().strftime(
    "%Y-%m-%d"
)

transformer_resume_df[
    "processing_version"
] = (
    "SkillBridge_Task3_Resume_Transformer_v1"
)

# ------------------------------------------------------------
# 10. ZERO-SKILL CHECK
# ------------------------------------------------------------

zero_skill_count = (
    transformer_resume_df[
        "transformer_candidate_skills"
    ]
    .apply(
        lambda x:
        len(x) if isinstance(x, list) else 0
    )
    .eq(0)
    .sum()
)

total_candidates = len(
    transformer_resume_df
)

zero_skill_rate = (
    zero_skill_count /
    total_candidates
    if total_candidates
    else 0
)

print("\n==============================================")
print("TRANSFORMER RESULTS")
print("==============================================")

print(
    "Total candidates:",
    total_candidates
)

print(
    "Candidates with extracted skills:",
    total_candidates - zero_skill_count
)

print(
    "Zero-skill candidates:",
    zero_skill_count
)

print(
    "Zero-skill rate:",
    round(
        zero_skill_rate * 100,
        2
    ),
    "%"
)

# ------------------------------------------------------------
# 11. SAVE OUTPUT
# ------------------------------------------------------------

output_file = (
    "39_Transformer_Resume_Skill_Extraction.xlsx"
)

transformer_resume_df.to_excel(
    output_file,
    index=False
)

print(
    "\n💾 Saved:",
    output_file
)

# ------------------------------------------------------------
# 12. DOWNLOAD
# ------------------------------------------------------------

files.download(
    output_file
)

print(
    "\n⬇️ Download started."
)

print("\n==============================================")
print("✅ TRANSFORMER RESUME EXTRACTION COMPLETED")
print("==============================================")
```

```python
# ============================================================
# CELL 62 — RESUME METHOD COMPARISON
# ============================================================

import pandas as pd
import ast
import re
from google.colab import files

print("==============================================")
print("RESUME EXTRACTION — 3 METHOD COMPARISON")
print("==============================================")

# ------------------------------------------------------------
# 1. CHECK REQUIRED DATA
# ------------------------------------------------------------

if "resume_benchmark" not in globals():
    raise ValueError(
        "❌ resume_benchmark not found. "
        "Run the verified manual resume benchmark cell first."
    )

if "resume_semantic_df" not in globals():
    raise ValueError(
        "❌ resume_semantic_df not found. "
        "Run the semantic resume extraction cell first."
    )

if "transformer_resume_df" not in globals():
    raise ValueError(
        "❌ transformer_resume_df not found. "
        "Run the transformer resume extraction cell first."
    )

gold_df = resume_benchmark.copy()
semantic_df = resume_semantic_df.copy()
trans_df = transformer_resume_df.copy()

# ------------------------------------------------------------
# 2. FLEXIBLE SKILL PARSER
# ------------------------------------------------------------

def parse_skills(value):

    if value is None:
        return set()

    if isinstance(value, (list, tuple, set)):
        return {
            str(x).strip().lower()
            for x in value
            if str(x).strip()
        }

    text = str(value).strip()

    if not text or text.lower() in [
        "nan",
        "none",
        "[]"
    ]:
        return set()

    # Try Python-list format
    try:
        parsed = ast.literal_eval(text)

        if isinstance(parsed, (list, tuple, set)):
            return {
                str(x).strip().lower()
                for x in parsed
                if str(x).strip()
            }

    except:
        pass

    # Otherwise split common formats
    parts = re.split(
        r"[,;|]",
        text
    )

    return {
        p.strip().lower()
        for p in parts
        if p.strip()
    }

# ------------------------------------------------------------
# 3. IDENTIFY BASELINE COLUMN
# ------------------------------------------------------------

baseline_candidates = [
    "candidate_skills",
    "extracted_skills",
    "normalized_skills",
    "predicted_skills"
]

baseline_column = None

for col in baseline_candidates:

    if col in gold_df.columns:
        baseline_column = col
        break

if baseline_column is None:

    for col in baseline_candidates:

        if col in semantic_df.columns:
            baseline_column = col
            break

if baseline_column is None:
    raise ValueError(
        "❌ Baseline candidate skill column not found."
    )

print(
    "Baseline column:",
    baseline_column
)

# ------------------------------------------------------------
# 4. PREPARE DATAFRAMES
# ------------------------------------------------------------

gold_df = gold_df[
    [
        "candidateId",
        "gold_candidate_skills"
    ]
].copy()

gold_df["gold_set"] = (
    gold_df["gold_candidate_skills"]
    .apply(parse_skills)
)

# Baseline
if baseline_column in semantic_df.columns:

    baseline_df = semantic_df[
        [
            "candidateId",
            baseline_column
        ]
    ].copy()

else:

    baseline_df = gold_df[
        ["candidateId"]
    ].copy()

    baseline_df[
        baseline_column
    ] = [[] for _ in range(
        len(baseline_df)
    )]

baseline_df["baseline_set"] = (
    baseline_df[baseline_column]
    .apply(parse_skills)
)

# Semantic
semantic_df = semantic_df[
    [
        "candidateId",
        "semantic_candidate_skills"
    ]
].copy()

semantic_df["semantic_set"] = (
    semantic_df[
        "semantic_candidate_skills"
    ]
    .apply(parse_skills)
)

# Transformer
trans_df = trans_df[
    [
        "candidateId",
        "transformer_candidate_skills"
    ]
].copy()

trans_df["transformer_set"] = (
    trans_df[
        "transformer_candidate_skills"
    ]
    .apply(parse_skills)
)

# ------------------------------------------------------------
# 5. MERGE
# ------------------------------------------------------------

comparison_df = gold_df.merge(
    baseline_df[
        ["candidateId", "baseline_set"]
    ],
    on="candidateId",
    how="left"
)

comparison_df = comparison_df.merge(
    semantic_df[
        ["candidateId", "semantic_set"]
    ],
    on="candidateId",
    how="left"
)

comparison_df = comparison_df.merge(
    trans_df[
        ["candidateId", "transformer_set"]
    ],
    on="candidateId",
    how="left"
)

for col in [
    "baseline_set",
    "semantic_set",
    "transformer_set"
]:

    comparison_df[col] = comparison_df[col].apply(
        lambda x:
        x if isinstance(x, set)
        else set()
    )

# ------------------------------------------------------------
# 6. METRIC FUNCTION
# ------------------------------------------------------------

def calculate_metrics(
    gold,
    predicted
):

    tp = len(
        gold.intersection(predicted)
    )

    fp = len(
        predicted.difference(gold)
    )

    fn = len(
        gold.difference(predicted)
    )

    precision = (
        tp / (tp + fp)
        if (tp + fp)
        else 0
    )

    recall = (
        tp / (tp + fn)
        if (tp + fn)
        else 0
    )

    f1 = (
        2 * precision * recall /
        (precision + recall)
        if (precision + recall)
        else 0
    )

    return (
        tp,
        fp,
        fn,
        precision,
        recall,
        f1
    )

# ------------------------------------------------------------
# 7. CALCULATE ALL METHODS
# ------------------------------------------------------------

method_columns = {
    "Baseline Dictionary/Regex":
        "baseline_set",

    "Sentence Transformer Semantic":
        "semantic_set",

    "Transformer NER":
        "transformer_set"
}

overall_results = []

for method, column in method_columns.items():

    TP = FP = FN = 0

    for _, row in comparison_df.iterrows():

        tp, fp, fn, _, _, _ = (
            calculate_metrics(
                row["gold_set"],
                row[column]
            )
        )

        TP += tp
        FP += fp
        FN += fn

    precision = (
        TP / (TP + FP)
        if (TP + FP)
        else 0
    )

    recall = (
        TP / (TP + FN)
        if (TP + FN)
        else 0
    )

    f1 = (
        2 * precision * recall /
        (precision + recall)
        if (precision + recall)
        else 0
    )

    overall_results.append({

        "method": method,

        "TP": TP,

        "FP": FP,

        "FN": FN,

        "Precision": round(
            precision, 4
        ),

        "Recall": round(
            recall, 4
        ),

        "F1": round(
            f1, 4
        ),

        "candidate_count":
            len(comparison_df)

    })

overall_metrics_df = pd.DataFrame(
    overall_results
)

# ------------------------------------------------------------
# 8. ZERO-SKILL ANALYSIS
# ------------------------------------------------------------

zero_skill_results = []

for method, column in method_columns.items():

    zero_count = (
        comparison_df[column]
        .apply(len)
        .eq(0)
        .sum()
    )

    total = len(
        comparison_df
    )

    zero_skill_results.append({

        "method": method,

        "total_candidates": total,

        "zero_skill_candidates":
            zero_count,

        "zero_skill_rate":
            round(
                zero_count / total,
                4
            )
            if total
            else 0

    })

zero_skill_df = pd.DataFrame(
    zero_skill_results
)

# ------------------------------------------------------------
# 9. CANDIDATE-LEVEL RESULTS
# ------------------------------------------------------------

candidate_results = []

for _, row in comparison_df.iterrows():

    for method, column in method_columns.items():

        tp, fp, fn, precision, recall, f1 = (
            calculate_metrics(
                row["gold_set"],
                row[column]
            )
        )

        candidate_results.append({

            "candidateId":
                row["candidateId"],

            "method":
                method,

            "TP":
                tp,

            "FP":
                fp,

            "FN":
                fn,

            "Precision":
                round(
                    precision, 4
                ),

            "Recall":
                round(
                    recall, 4
                ),

            "F1":
                round(
                    f1, 4
                )

        })

candidate_metrics_df = pd.DataFrame(
    candidate_results
)

# ------------------------------------------------------------
# 10. DISPLAY
# ------------------------------------------------------------

print("\nOVERALL RESULTS")
display(
    overall_metrics_df
)

print("\nZERO-SKILL RESULTS")
display(
    zero_skill_df
)

print("\nCANDIDATE-LEVEL RESULTS")
display(
    candidate_metrics_df.head(20)
)

# ------------------------------------------------------------
# 11. SAVE EXCEL
# ------------------------------------------------------------

output_file = (
    "40_Task3_Resume_Three_Method_Comparison.xlsx"
)

with pd.ExcelWriter(
    output_file,
    engine="openpyxl"
) as writer:

    overall_metrics_df.to_excel(
        writer,
        sheet_name="Overall_Metrics",
        index=False
    )

    zero_skill_df.to_excel(
        writer,
        sheet_name="Zero_Skill_Analysis",
        index=False
    )

    candidate_metrics_df.to_excel(
        writer,
        sheet_name="Candidate_Metrics",
        index=False
    )

    comparison_df[
        [
            "candidateId",
            "gold_candidate_skills"
        ]
    ].to_excel(
        writer,
        sheet_name="Gold_Benchmark",
        index=False
    )

print(
    "\n💾 Saved:",
    output_file
)

# ------------------------------------------------------------
# 12. DOWNLOAD
# ------------------------------------------------------------

files.download(
    output_file
)

print("\n==============================================")
print("✅ THREE-METHOD RESUME EVALUATION COMPLETED")
print("==============================================")
```

```python
# ============================================================
# CELL 64 — RESUME CONFIDENCE / SIMILARITY CALIBRATION
# ============================================================

import pandas as pd
import ast
import re
import numpy as np
from google.colab import files

print("==============================================")
print("RESUME CONFIDENCE CALIBRATION")
print("==============================================")

# ------------------------------------------------------------
# 1. CHECK DATA
# ------------------------------------------------------------

if "resume_semantic_df" not in globals():
    raise ValueError(
        "❌ resume_semantic_df not found. "
        "Run the semantic resume extraction cell first."
    )

if "resume_benchmark" not in globals():
    raise ValueError(
        "❌ resume_benchmark not found. "
        "Run the verified manual resume benchmark first."
    )

semantic_df = resume_semantic_df.copy()
gold_df = resume_benchmark.copy()

# ------------------------------------------------------------
# 2. SKILL PARSER
# ------------------------------------------------------------

def parse_skills(value):

    if value is None:
        return set()

    if isinstance(value, (list, tuple, set)):
        return {
            str(x).strip().lower()
            for x in value
            if str(x).strip()
        }

    text = str(value).strip()

    if not text or text.lower() in [
        "nan",
        "none",
        "[]"
    ]:
        return set()

    try:
        parsed = ast.literal_eval(text)

        if isinstance(
            parsed,
            (list, tuple, set)
        ):
            return {
                str(x).strip().lower()
                for x in parsed
                if str(x).strip()
            }

    except:
        pass

    return {
        x.strip().lower()
        for x in re.split(
            r"[,;|]",
            text
        )
        if x.strip()
    }

# ------------------------------------------------------------
# 3. FIND CONFIDENCE COLUMNS
# ------------------------------------------------------------

confidence_column = None

for col in [
    "semantic_confidence",
    "candidate_skill_similarity_scores"
]:

    if col in semantic_df.columns:
        confidence_column = col
        break

if confidence_column is None:
    print(
        "⚠️ No per-skill semantic confidence column found."
    )

# ------------------------------------------------------------
# 4. PREPARE GOLD SKILLS
# ------------------------------------------------------------

gold_df = gold_df[
    [
        "candidateId",
        "gold_candidate_skills"
    ]
].copy()

gold_df["gold_set"] = (
    gold_df[
        "gold_candidate_skills"
    ]
    .apply(parse_skills)
)

# ------------------------------------------------------------
# 5. PREPARE SEMANTIC SKILLS
# ------------------------------------------------------------

required_columns = [
    "candidateId",
    "semantic_candidate_skills"
]

for col in required_columns:

    if col not in semantic_df.columns:
        raise ValueError(
            f"❌ Required column missing: {col}"
        )

semantic_df["semantic_set"] = (
    semantic_df[
        "semantic_candidate_skills"
    ]
    .apply(parse_skills)
)

# ------------------------------------------------------------
# 6. MERGE
# ------------------------------------------------------------

calibration_df = gold_df.merge(
    semantic_df,
    on="candidateId",
    how="inner"
)

print(
    "Candidates evaluated:",
    len(calibration_df)
)

# ------------------------------------------------------------
# 7. CREATE SKILL-LEVEL CONFIDENCE RECORDS
# ------------------------------------------------------------

records = []

for _, row in calibration_df.iterrows():

    candidate_id = row[
        "candidateId"
    ]

    gold = row[
        "gold_set"
    ]

    predicted = row[
        "semantic_set"
    ]

    # --------------------------------------------------------
    # Confidence values
    # --------------------------------------------------------

    confidence_values = {}

    if confidence_column is not None:

        raw_conf = row[
            confidence_column
        ]

        if isinstance(
            raw_conf,
            dict
        ):
            confidence_values = raw_conf

        elif isinstance(
            raw_conf,
            list
        ):

            predicted_list = list(
                predicted
            )

            for i, skill in enumerate(
                predicted_list
            ):

                if i < len(raw_conf):

                    try:
                        confidence_values[
                            skill
                        ] = float(
                            raw_conf[i]
                        )
                    except:
                        pass

        elif isinstance(
            raw_conf,
            str
        ):

            try:

                parsed = ast.literal_eval(
                    raw_conf
                )

                if isinstance(
                    parsed,
                    dict
                ):
                    confidence_values = parsed

            except:
                pass

    # --------------------------------------------------------
    # Create records
    # --------------------------------------------------------

    for skill in predicted:

        score = confidence_values.get(
            skill,
            np.nan
        )

        status = (
            "CORRECT"
            if skill in gold
            else "INCORRECT"
        )

        records.append({

            "candidateId":
                candidate_id,

            "skill":
                skill,

            "confidence_similarity":
                score,

            "match_status":
                status

        })

confidence_detail_df = pd.DataFrame(
    records
)

# ------------------------------------------------------------
# 8. REMOVE INVALID SCORES
# ------------------------------------------------------------

if len(confidence_detail_df) > 0:

    confidence_detail_df[
        "confidence_similarity"
    ] = pd.to_numeric(
        confidence_detail_df[
            "confidence_similarity"
        ],
        errors="coerce"
    )

    valid_confidence_df = (
        confidence_detail_df
        .dropna(
            subset=[
                "confidence_similarity"
            ]
        )
        .copy()
    )

else:

    valid_confidence_df = pd.DataFrame()

# ------------------------------------------------------------
# 9. CONFIDENCE BUCKETS
# ------------------------------------------------------------

if len(valid_confidence_df) > 0:

    valid_confidence_df[
        "confidence_band"
    ] = pd.cut(
        valid_confidence_df[
            "confidence_similarity"
        ],
        bins=[
            -np.inf,
            0.55,
            0.65,
            0.75,
            0.85,
            np.inf
        ],
        labels=[
            "<0.55",
            "0.55–0.65",
            "0.65–0.75",
            "0.75–0.85",
            ">=0.85"
        ]
    )

    calibration_summary_df = (
        valid_confidence_df
        .groupby(
            [
                "confidence_band",
                "match_status"
            ],
            observed=False
        )
        .size()
        .reset_index(
            name="count"
        )
    )

else:

    calibration_summary_df = pd.DataFrame(
        columns=[
            "confidence_band",
            "match_status",
            "count"
        ]
    )

# ------------------------------------------------------------
# 10. OVERALL CONFIDENCE SUMMARY
# ------------------------------------------------------------

if len(valid_confidence_df) > 0:

    overall_confidence_df = pd.DataFrame({

        "metric": [

            "Total predicted skill matches",

            "Valid similarity scores",

            "Mean similarity",

            "Mean similarity - correct matches",

            "Mean similarity - incorrect matches"

        ],

        "value": [

            len(confidence_detail_df),

            len(valid_confidence_df),

            valid_confidence_df[
                "confidence_similarity"
            ].mean(),

            valid_confidence_df.loc[
                valid_confidence_df[
                    "match_status"
                ] == "CORRECT",
                "confidence_similarity"
            ].mean(),

            valid_confidence_df.loc[
                valid_confidence_df[
                    "match_status"
                ] == "INCORRECT",
                "confidence_similarity"
            ].mean()

        ]

    })

else:

    overall_confidence_df = pd.DataFrame({

        "metric": [
            "Total predicted skill matches",
            "Valid similarity scores",
            "Mean similarity",
            "Mean similarity - correct matches",
            "Mean similarity - incorrect matches"
        ],

        "value": [
            len(confidence_detail_df),
            0,
            np.nan,
            np.nan,
            np.nan
        ]

    })

# ------------------------------------------------------------
# 11. CONFIDENCE DEFINITION
# ------------------------------------------------------------

definition_df = pd.DataFrame({

    "field": [

        "confidence_similarity",

        "score_type",

        "interpretation",

        "is_probability",

        "calibration_reference"

    ],

    "definition": [

        "Cosine similarity between resume text representation and skill representation",

        "Semantic similarity score",

        "Higher values indicate greater semantic similarity between the candidate profile and the skill representation",

        "No",

        "Manually validated gold_candidate_skills"

    ]

})

# ------------------------------------------------------------
# 12. DISPLAY
# ------------------------------------------------------------

print("\nOVERALL CONFIDENCE")
display(
    overall_confidence_df
)

print("\nCONFIDENCE BANDS")
display(
    calibration_summary_df
)

print("\nCONFIDENCE DEFINITION")
display(
    definition_df
)

# ------------------------------------------------------------
# 13. SAVE
# ------------------------------------------------------------

output_file = (
    "42_Resume_Confidence_Calibration.xlsx"
)

with pd.ExcelWriter(
    output_file,
    engine="openpyxl"
) as writer:

    overall_confidence_df.to_excel(
        writer,
        sheet_name="Overall_Confidence",
        index=False
    )

    calibration_summary_df.to_excel(
        writer,
        sheet_name="Confidence_Bands",
        index=False
    )

    confidence_detail_df.to_excel(
        writer,
        sheet_name="Skill_Level_Detail",
        index=False
    )

    definition_df.to_excel(
        writer,
        sheet_name="Score_Definition",
        index=False
    )

print(
    "\n💾 Saved:",
    output_file
)

# ------------------------------------------------------------
# 14. DOWNLOAD
# ------------------------------------------------------------

files.download(
    output_file
)

print("\n==============================================")
print("✅ CONFIDENCE CALIBRATION COMPLETED")
print("==============================================")
```

```python
# ============================================================
# CELL 65 — FINAL JOB + RESUME G2-A / G2-B INTEGRATION
# ============================================================

import pandas as pd
import ast
import re
from datetime import datetime
from google.colab import files

print("==============================================")
print("FINAL JOB + RESUME INTEGRATION")
print("==============================================")

# ------------------------------------------------------------
# 1. FIND JOB DATASET
# ------------------------------------------------------------

job_source = None

for name in [
    "final_unified_dataset",
    "transformer_df",
    "semantic_df",
    "evaluation_df",
    "df"
]:
    if name in globals():
        obj = globals()[name]

        if isinstance(obj, pd.DataFrame):
            job_source = obj.copy()
            print("Job dataset found:", name)
            break

if job_source is None:
    raise ValueError(
        "❌ No job dataset found. "
        "Run the job NLP pipeline first."
    )

# ------------------------------------------------------------
# 2. FIND RESUME DATASET
# ------------------------------------------------------------

resume_source = None

for name in [
    "final_resume_gap_df",
    "resume_semantic_df",
    "resume_benchmark",
    "transformer_resume_df"
]:
    if name in globals():
        obj = globals()[name]

        if isinstance(obj, pd.DataFrame):
            resume_source = obj.copy()
            print("Resume dataset found:", name)
            break

if resume_source is None:
    raise ValueError(
        "❌ No resume dataset found. "
        "Run the resume pipeline first."
    )

# ------------------------------------------------------------
# 3. SKILL PARSER
# ------------------------------------------------------------

def parse_skills(value):

    if value is None:
        return []

    if isinstance(value, (list, tuple, set)):
        return sorted(
            set(
                str(x).strip()
                for x in value
                if str(x).strip()
            )
        )

    text = str(value).strip()

    if not text or text.lower() in [
        "nan",
        "none",
        "[]"
    ]:
        return []

    try:
        parsed = ast.literal_eval(text)

        if isinstance(
            parsed,
            (list, tuple, set)
        ):
            return sorted(
                set(
                    str(x).strip()
                    for x in parsed
                    if str(x).strip()
                )
            )

    except:
        pass

    return sorted(
        set(
            x.strip()
            for x in re.split(
                r"[,;|]",
                text
            )
            if x.strip()
        )
    )

# ------------------------------------------------------------
# 4. HELPER FOR COLUMN DETECTION
# ------------------------------------------------------------

def first_existing(df, candidates):

    for col in candidates:
        if col in df.columns:
            return col

    return None

# ------------------------------------------------------------
# 5. STANDARDIZE JOB SCHEMA
# ------------------------------------------------------------

job_output = pd.DataFrame()

# Job ID
job_id_col = first_existing(
    job_source,
    [
        "jobId",
        "Job ID",
        "job_id",
        "id"
    ]
)

if job_id_col:
    job_output["jobId"] = (
        job_source[job_id_col]
        .astype(str)
    )
else:
    job_output["jobId"] = [
        f"JOB_{i+1:06d}"
        for i in range(
            len(job_source)
        )
    ]

# Title
title_col = first_existing(
    job_source,
    [
        "title",
        "job_title",
        "jobTitle"
    ]
)

job_output["title"] = (
    job_source[title_col].astype(str)
    if title_col
    else ""
)

# Company
company_col = first_existing(
    job_source,
    [
        "companyName",
        "company",
        "employer",
        "employerName"
    ]
)

job_output["companyName"] = (
    job_source[company_col].astype(str)
    if company_col
    else ""
)

# Country
country_col = first_existing(
    job_source,
    [
        "country",
        "Country"
    ]
)

job_output["country"] = (
    job_source[country_col].astype(str)
    if country_col
    else ""
)

# Location
location_col = first_existing(
    job_source,
    [
        "location",
        "job_location",
        "city"
    ]
)

job_output["location"] = (
    job_source[location_col].astype(str)
    if location_col
    else ""
)

# Description
description_col = first_existing(
    job_source,
    [
        "jobDescription",
        "job_description",
        "description"
    ]
)

job_output["jobDescription"] = (
    job_source[description_col].astype(str)
    if description_col
    else ""
)

# NLP text
nlp_col = first_existing(
    job_source,
    [
        "nlp_text",
        "processed_text"
    ]
)

if nlp_col:
    job_output["nlp_text"] = (
        job_source[nlp_col].astype(str)
    )
else:
    job_output["nlp_text"] = (
        job_output["title"].fillna("")
        + " "
        + job_output["jobDescription"].fillna("")
    )

# ------------------------------------------------------------
# 6. JOB SKILLS
# ------------------------------------------------------------

for output_col, candidates in {

    "extracted_skills": [
        "extracted_skills",
        "baseline_skills"
    ],

    "normalized_skills": [
        "normalized_skills"
    ],

    "semantic_skills": [
        "semantic_skills"
    ],

    "skill_categories": [
        "skill_categories"
    ],

    "skill_confidence": [
        "skill_confidence",
        "semantic_confidence"
    ],

    "confidence_score": [
        "confidence_score",
        "semantic_confidence_score"
    ]

}.items():

    source_col = first_existing(
        job_source,
        candidates
    )

    if source_col:
        job_output[output_col] = (
            job_source[source_col]
            .apply(parse_skills)
        )
    else:
        job_output[output_col] = [
            []
            for _ in range(
                len(job_source)
            )
        ]

# ------------------------------------------------------------
# 7. JOB METADATA
# ------------------------------------------------------------

job_output[
    "nlp_model_version"
] = (
    "SkillBridge_Task3_NLP_"
    "Baseline_Semantic_Transformer"
)

job_output[
    "processing_version"
] = (
    "SkillBridge_Task3_Final_v1"
)

job_output[
    "processing_date"
] = datetime.now().strftime(
    "%Y-%m-%d"
)

job_output[
    "source_type"
] = "Job Advertisement"

# ------------------------------------------------------------
# 8. STANDARDIZE RESUME SCHEMA
# ------------------------------------------------------------

resume_output = pd.DataFrame()

candidate_id_col = first_existing(
    resume_source,
    [
        "candidateId",
        "candidate_id",
        "Candidate ID"
    ]
)

if candidate_id_col:

    resume_output["candidateId"] = (
        resume_source[
            candidate_id_col
        ].astype(str)
    )

else:

    resume_output["candidateId"] = [
        f"CAND_{i+1:04d}"
        for i in range(
            len(resume_source)
        )
    ]

# Resume text
resume_text_col = first_existing(
    resume_source,
    [
        "candidate_profile_text",
        "resume_text",
        "profile_text",
        "text"
    ]
)

resume_output[
    "candidate_profile_text"
] = (
    resume_source[
        resume_text_col
    ].astype(str)
    if resume_text_col
    else ""
)

# Validated skills
gold_resume_col = first_existing(
    resume_source,
    [
        "validated_candidate_skills",
        "gold_candidate_skills"
    ]
)

resume_output[
    "validated_candidate_skills"
] = (
    resume_source[
        gold_resume_col
    ].apply(parse_skills)
    if gold_resume_col
    else [[] for _ in range(
        len(resume_source)
    )]
)

# Baseline skills
baseline_resume_col = first_existing(
    resume_source,
    [
        "candidate_skills",
        "extracted_skills"
    ]
)

resume_output[
    "candidate_skills"
] = (
    resume_source[
        baseline_resume_col
    ].apply(parse_skills)
    if baseline_resume_col
    else [[] for _ in range(
        len(resume_source)
    )]
)

# Semantic skills
semantic_resume_col = first_existing(
    resume_source,
    [
        "candidate_semantic_skills",
        "semantic_candidate_skills"
    ]
)

resume_output[
    "candidate_semantic_skills"
] = (
    resume_source[
        semantic_resume_col
    ].apply(parse_skills)
    if semantic_resume_col
    else [[] for _ in range(
        len(resume_source)
    )]
)

# ------------------------------------------------------------
# 9. RESUME GAP FIELDS
# ------------------------------------------------------------

gap_columns = [
    "candidate_skill_similarity_scores",
    "candidate_confidence_score",
    "required_skill_count",
    "candidate_skill_count",
    "missing_skill_count",
    "missing_skills",
    "weighted_gap_score",
    "weighted_gap_percentage",
    "traceability_record_count"
]

for col in gap_columns:

    if col in resume_source.columns:

        resume_output[col] = (
            resume_source[col]
        )

    else:

        if col in [
            "missing_skills"
        ]:

            resume_output[col] = [
                []
                for _ in range(
                    len(resume_source)
                )
            ]

        else:

            resume_output[col] = 0

# ------------------------------------------------------------
# 10. RESUME METADATA
# ------------------------------------------------------------

resume_output[
    "nlp_model_version"
] = (
    "SkillBridge_Task3_Resume_"
    "Baseline_Semantic_Transformer"
)

resume_output[
    "processing_version"
] = (
    "SkillBridge_Task3_Final_v1"
)

resume_output[
    "processing_date"
] = datetime.now().strftime(
    "%Y-%m-%d"
)

resume_output[
    "source_type"
] = "Candidate Resume"

resume_output[
    "benchmark_status"
] = "MANUALLY_VALIDATED"

# ------------------------------------------------------------
# 11. INTEGRATION SUMMARY
# ------------------------------------------------------------

integration_summary = pd.DataFrame({

    "dataset": [
        "Job Dataset",
        "Resume Dataset"
    ],

    "record_count": [
        len(job_output),
        len(resume_output)
    ],

    "unique_id_count": [
        job_output[
            "jobId"
        ].nunique(),

        resume_output[
            "candidateId"
        ].nunique()
    ],

    "empty_skill_records": [

        job_output[
            "normalized_skills"
        ].apply(len).eq(0).sum(),

        resume_output[
            "candidate_semantic_skills"
        ].apply(len).eq(0).sum()
    ]

})

print("\n==============================================")
print("INTEGRATION SUMMARY")
print("==============================================")

display(
    integration_summary
)

# ------------------------------------------------------------
# 12. SAVE FINAL JOB DATASET
# ------------------------------------------------------------

job_file = (
    "43_FINAL_G2A_Job_NLP_Dataset.xlsx"
)

job_csv = (
    "43_FINAL_G2A_Job_NLP_Dataset.csv"
)

job_output.to_excel(
    job_file,
    index=False
)

job_output.to_csv(
    job_csv,
    index=False
)

# ------------------------------------------------------------
# 13. SAVE FINAL RESUME DATASET
# ------------------------------------------------------------

resume_file = (
    "43_FINAL_G2B_Resume_Skill_Gap_Dataset.xlsx"
)

resume_csv = (
    "43_FINAL_G2B_Resume_Skill_Gap_Dataset.csv"
)

resume_output.to_excel(
    resume_file,
    index=False
)

resume_output.to_csv(
    resume_csv,
    index=False
)

# ------------------------------------------------------------
# 14. SAVE INTEGRATION SUMMARY
# ------------------------------------------------------------

summary_file = (
    "43_G2_Integration_Summary.xlsx"
)

integration_summary.to_excel(
    summary_file,
    index=False
)

# ------------------------------------------------------------
# 15. DOWNLOAD ALL
# ------------------------------------------------------------

print("\n💾 Files created:")

print(job_file)
print(job_csv)
print(resume_file)
print(resume_csv)
print(summary_file)

files.download(job_file)
files.download(job_csv)
files.download(resume_file)
files.download(resume_csv)
files.download(summary_file)

print("\n==============================================")
print("✅ G2-A / G2-B INTEGRATION COMPLETED")
print("==============================================")
```

```python
# ============================================================
# CELL 65 — FINAL JOB + RESUME G2-A / G2-B INTEGRATION
# ============================================================

import pandas as pd
import ast
import re
from datetime import datetime
from google.colab import files

print("==============================================")
print("FINAL JOB + RESUME INTEGRATION")
print("==============================================")

# ------------------------------------------------------------
# 1. FIND JOB DATASET
# ------------------------------------------------------------

job_source = None

for name in [
    "final_unified_dataset",
    "transformer_df",
    "semantic_df",
    "evaluation_df",
    "df"
]:
    if name in globals():
        obj = globals()[name]

        if isinstance(obj, pd.DataFrame):
            job_source = obj.copy()
            print("Job dataset found:", name)
            break

if job_source is None:
    raise ValueError(
        "❌ No job dataset found. "
        "Run the job NLP pipeline first."
    )

# ------------------------------------------------------------
# 2. FIND RESUME DATASET
# ------------------------------------------------------------

resume_source = None

for name in [
    "final_resume_gap_df",
    "resume_semantic_df",
    "resume_benchmark",
    "transformer_resume_df"
]:
    if name in globals():
        obj = globals()[name]

        if isinstance(obj, pd.DataFrame):
            resume_source = obj.copy()
            print("Resume dataset found:", name)
            break

if resume_source is None:
    raise ValueError(
        "❌ No resume dataset found. "
        "Run the resume pipeline first."
    )

# ------------------------------------------------------------
# 3. SKILL PARSER
# ------------------------------------------------------------

def parse_skills(value):

    if value is None:
        return []

    if isinstance(value, (list, tuple, set)):
        return sorted(
            set(
                str(x).strip()
                for x in value
                if str(x).strip()
            )
        )

    text = str(value).strip()

    if not text or text.lower() in [
        "nan",
        "none",
        "[]"
    ]:
        return []

    try:
        parsed = ast.literal_eval(text)

        if isinstance(
            parsed,
            (list, tuple, set)
        ):
            return sorted(
                set(
                    str(x).strip()
                    for x in parsed
                    if str(x).strip()
                )
            )

    except:
        pass

    return sorted(
        set(
            x.strip()
            for x in re.split(
                r"[,;|]",
                text
            )
            if x.strip()
        )
    )

# ------------------------------------------------------------
# 4. HELPER FOR COLUMN DETECTION
# ------------------------------------------------------------

def first_existing(df, candidates):

    for col in candidates:
        if col in df.columns:
            return col

    return None

# ------------------------------------------------------------
# 5. STANDARDIZE JOB SCHEMA
# ------------------------------------------------------------

job_output = pd.DataFrame()

# Job ID
job_id_col = first_existing(
    job_source,
    [
        "jobId",
        "Job ID",
        "job_id",
        "id"
    ]
)

if job_id_col:
    job_output["jobId"] = (
        job_source[job_id_col]
        .astype(str)
    )
else:
    job_output["jobId"] = [
        f"JOB_{i+1:06d}"
        for i in range(
            len(job_source)
        )
    ]

# Title
title_col = first_existing(
    job_source,
    [
        "title",
        "job_title",
        "jobTitle"
    ]
)

job_output["title"] = (
    job_source[title_col].astype(str)
    if title_col
    else ""
)

# Company
company_col = first_existing(
    job_source,
    [
        "companyName",
        "company",
        "employer",
        "employerName"
    ]
)

job_output["companyName"] = (
    job_source[company_col].astype(str)
    if company_col
    else ""
)

# Country
country_col = first_existing(
    job_source,
    [
        "country",
        "Country"
    ]
)

job_output["country"] = (
    job_source[country_col].astype(str)
    if country_col
    else ""
)

# Location
location_col = first_existing(
    job_source,
    [
        "location",
        "job_location",
        "city"
    ]
)

job_output["location"] = (
    job_source[location_col].astype(str)
    if location_col
    else ""
)

# Description
description_col = first_existing(
    job_source,
    [
        "jobDescription",
        "job_description",
        "description"
    ]
)

job_output["jobDescription"] = (
    job_source[description_col].astype(str)
    if description_col
    else ""
)

# NLP text
nlp_col = first_existing(
    job_source,
    [
        "nlp_text",
        "processed_text"
    ]
)

if nlp_col:
    job_output["nlp_text"] = (
        job_source[nlp_col].astype(str)
    )
else:
    job_output["nlp_text"] = (
        job_output["title"].fillna("")
        + " "
        + job_output["jobDescription"].fillna("")
    )

# ------------------------------------------------------------
# 6. JOB SKILLS
# ------------------------------------------------------------

for output_col, candidates in {

    "extracted_skills": [
        "extracted_skills",
        "baseline_skills"
    ],

    "normalized_skills": [
        "normalized_skills"
    ],

    "semantic_skills": [
        "semantic_skills"
    ],

    "skill_categories": [
        "skill_categories"
    ],

    "skill_confidence": [
        "skill_confidence",
        "semantic_confidence"
    ],

    "confidence_score": [
        "confidence_score",
        "semantic_confidence_score"
    ]

}.items():

    source_col = first_existing(
        job_source,
        candidates
    )

    if source_col:
        job_output[output_col] = (
            job_source[source_col]
            .apply(parse_skills)
        )
    else:
        job_output[output_col] = [
            []
            for _ in range(
                len(job_source)
            )
        ]

# ------------------------------------------------------------
# 7. JOB METADATA
# ------------------------------------------------------------

job_output[
    "nlp_model_version"
] = (
    "SkillBridge_Task3_NLP_"
    "Baseline_Semantic_Transformer"
)

job_output[
    "processing_version"
] = (
    "SkillBridge_Task3_Final_v1"
)

job_output[
    "processing_date"
] = datetime.now().strftime(
    "%Y-%m-%d"
)

job_output[
    "source_type"
] = "Job Advertisement"

# ------------------------------------------------------------
# 8. STANDARDIZE RESUME SCHEMA
# ------------------------------------------------------------

resume_output = pd.DataFrame()

candidate_id_col = first_existing(
    resume_source,
    [
        "candidateId",
        "candidate_id",
        "Candidate ID"
    ]
)

if candidate_id_col:

    resume_output["candidateId"] = (
        resume_source[
            candidate_id_col
        ].astype(str)
    )

else:

    resume_output["candidateId"] = [
        f"CAND_{i+1:04d}"
        for i in range(
            len(resume_source)
        )
    ]

# Resume text
resume_text_col = first_existing(
    resume_source,
    [
        "candidate_profile_text",
        "resume_text",
        "profile_text",
        "text"
    ]
)

resume_output[
    "candidate_profile_text"
] = (
    resume_source[
        resume_text_col
    ].astype(str)
    if resume_text_col
    else ""
)

# Validated skills
gold_resume_col = first_existing(
    resume_source,
    [
        "validated_candidate_skills",
        "gold_candidate_skills"
    ]
)

resume_output[
    "validated_candidate_skills"
] = (
    resume_source[
        gold_resume_col
    ].apply(parse_skills)
    if gold_resume_col
    else [[] for _ in range(
        len(resume_source)
    )]
)

# Baseline skills
baseline_resume_col = first_existing(
    resume_source,
    [
        "candidate_skills",
        "extracted_skills"
    ]
)

resume_output[
    "candidate_skills"
] = (
    resume_source[
        baseline_resume_col
    ].apply(parse_skills)
    if baseline_resume_col
    else [[] for _ in range(
        len(resume_source)
    )]
)

# Semantic skills
semantic_resume_col = first_existing(
    resume_source,
    [
        "candidate_semantic_skills",
        "semantic_candidate_skills"
    ]
)

resume_output[
    "candidate_semantic_skills"
] = (
    resume_source[
        semantic_resume_col
    ].apply(parse_skills)
    if semantic_resume_col
    else [[] for _ in range(
        len(resume_source)
    )]
)

# ------------------------------------------------------------
# 9. RESUME GAP FIELDS
# ------------------------------------------------------------

gap_columns = [
    "candidate_skill_similarity_scores",
    "candidate_confidence_score",
    "required_skill_count",
    "candidate_skill_count",
    "missing_skill_count",
    "missing_skills",
    "weighted_gap_score",
    "weighted_gap_percentage",
    "traceability_record_count"
]

for col in gap_columns:

    if col in resume_source.columns:

        resume_output[col] = (
            resume_source[col]
        )

    else:

        if col in [
            "missing_skills"
        ]:

            resume_output[col] = [
                []
                for _ in range(
                    len(resume_source)
                )
            ]

        else:

            resume_output[col] = 0

# ------------------------------------------------------------
# 10. RESUME METADATA
# ------------------------------------------------------------

resume_output[
    "nlp_model_version"
] = (
    "SkillBridge_Task3_Resume_"
    "Baseline_Semantic_Transformer"
)

resume_output[
    "processing_version"
] = (
    "SkillBridge_Task3_Final_v1"
)

resume_output[
    "processing_date"
] = datetime.now().strftime(
    "%Y-%m-%d"
)

resume_output[
    "source_type"
] = "Candidate Resume"

resume_output[
    "benchmark_status"
] = "MANUALLY_VALIDATED"

# ------------------------------------------------------------
# 11. INTEGRATION SUMMARY
# ------------------------------------------------------------

integration_summary = pd.DataFrame({

    "dataset": [
        "Job Dataset",
        "Resume Dataset"
    ],

    "record_count": [
        len(job_output),
        len(resume_output)
    ],

    "unique_id_count": [
        job_output[
            "jobId"
        ].nunique(),

        resume_output[
            "candidateId"
        ].nunique()
    ],

    "empty_skill_records": [

        job_output[
            "normalized_skills"
        ].apply(len).eq(0).sum(),

        resume_output[
            "candidate_semantic_skills"
        ].apply(len).eq(0).sum()
    ]

})

print("\n==============================================")
print("INTEGRATION SUMMARY")
print("==============================================")

display(
    integration_summary
)

# ------------------------------------------------------------
# 12. SAVE FINAL JOB DATASET
# ------------------------------------------------------------

job_file = (
    "43_FINAL_G2A_Job_NLP_Dataset.xlsx"
)

job_csv = (
    "43_FINAL_G2A_Job_NLP_Dataset.csv"
)

job_output.to_excel(
    job_file,
    index=False
)

job_output.to_csv(
    job_csv,
    index=False
)

# ------------------------------------------------------------
# 13. SAVE FINAL RESUME DATASET
# ------------------------------------------------------------

resume_file = (
    "43_FINAL_G2B_Resume_Skill_Gap_Dataset.xlsx"
)

resume_csv = (
    "43_FINAL_G2B_Resume_Skill_Gap_Dataset.csv"
)

resume_output.to_excel(
    resume_file,
    index=False
)

resume_output.to_csv(
    resume_csv,
    index=False
)

# ------------------------------------------------------------
# 14. SAVE INTEGRATION SUMMARY
# ------------------------------------------------------------

summary_file = (
    "43_G2_Integration_Summary.xlsx"
)

integration_summary.to_excel(
    summary_file,
    index=False
)

# ------------------------------------------------------------
# 15. DOWNLOAD ALL
# ------------------------------------------------------------

print("\n💾 Files created:")

print(job_file)
print(job_csv)
print(resume_file)
print(resume_csv)
print(summary_file)

files.download(job_file)
files.download(job_csv)
files.download(resume_file)
files.download(resume_csv)
files.download(summary_file)

print("\n==============================================")
print("✅ G2-A / G2-B INTEGRATION COMPLETED")
print("==============================================")
```

```python
# ============================================================
# CELL 67 — FINAL DATA QUALITY & INTEGRATION TESTS
# ============================================================

import pandas as pd
import numpy as np
import ast
import re
from google.colab import files

print("==============================================")
print("FINAL DATA QUALITY & INTEGRATION TESTS")
print("==============================================")

# ------------------------------------------------------------
# 1. CHECK DATASETS
# ------------------------------------------------------------

if "job_output" not in globals():
    raise ValueError(
        "❌ job_output not found. Run Cell 65 first."
    )

if "resume_output" not in globals():
    raise ValueError(
        "❌ resume_output not found. Run Cell 65 first."
    )

job_df = job_output.copy()
resume_df = resume_output.copy()

# ------------------------------------------------------------
# 2. REQUIRED FIELDS
# ------------------------------------------------------------

job_required = [
    "jobId",
    "title",
    "companyName",
    "country",
    "jobDescription",
    "nlp_text",
    "extracted_skills",
    "normalized_skills",
    "nlp_model_version",
    "processing_version",
    "processing_date"
]

resume_required = [
    "candidateId",
    "candidate_profile_text",
    "validated_candidate_skills",
    "candidate_skills",
    "candidate_semantic_skills",
    "required_skill_count",
    "candidate_skill_count",
    "missing_skill_count",
    "missing_skills",
    "weighted_gap_score",
    "weighted_gap_percentage",
    "nlp_model_version",
    "processing_version",
    "processing_date",
    "benchmark_status"
]

# ------------------------------------------------------------
# 3. TEST HELPER
# ------------------------------------------------------------

test_results = []

def add_test(
    test_name,
    dataset,
    passed,
    details
):

    test_results.append({

        "test_name":
            test_name,

        "dataset":
            dataset,

        "status":
            "PASS" if passed else "FAIL",

        "details":
            details

    })

# ------------------------------------------------------------
# 4. JOB SCHEMA TEST
# ------------------------------------------------------------

missing_job_fields = [
    x
    for x in job_required
    if x not in job_df.columns
]

add_test(
    "Job required fields",
    "G2-A",
    len(missing_job_fields) == 0,
    "Missing: "
    + (
        ", ".join(missing_job_fields)
        if missing_job_fields
        else "NONE"
    )
)

# ------------------------------------------------------------
# 5. RESUME SCHEMA TEST
# ------------------------------------------------------------

missing_resume_fields = [
    x
    for x in resume_required
    if x not in resume_df.columns
]

add_test(
    "Resume required fields",
    "G2-B",
    len(missing_resume_fields) == 0,
    "Missing: "
    + (
        ", ".join(missing_resume_fields)
        if missing_resume_fields
        else "NONE"
    )
)

# ------------------------------------------------------------
# 6. JOB ID TEST
# ------------------------------------------------------------

job_id_valid = (
    "jobId" in job_df.columns
    and
    job_df["jobId"]
    .notna()
    .all()
    and
    job_df["jobId"]
    .astype(str)
    .str.strip()
    .ne("")
    .all()
)

add_test(
    "Job IDs are non-empty",
    "G2-A",
    job_id_valid,
    f"Records checked: {len(job_df)}"
)

# ------------------------------------------------------------
# 7. JOB DUPLICATE TEST
# ------------------------------------------------------------

if "jobId" in job_df.columns:

    duplicate_jobs = (
        job_df["jobId"]
        .astype(str)
        .duplicated()
        .sum()
    )

else:

    duplicate_jobs = -1

add_test(
    "Duplicate Job IDs",
    "G2-A",
    duplicate_jobs == 0,
    f"Duplicate IDs: {duplicate_jobs}"
)

# ------------------------------------------------------------
# 8. JOB TITLE TEST
# ------------------------------------------------------------

if "title" in job_df.columns:

    empty_titles = (
        job_df["title"]
        .fillna("")
        .astype(str)
        .str.strip()
        .eq("")
        .sum()
    )

else:

    empty_titles = len(job_df)

add_test(
    "Job titles present",
    "G2-A",
    empty_titles == 0,
    f"Empty titles: {empty_titles}"
)

# ------------------------------------------------------------
# 9. JOB DESCRIPTION TEST
# ------------------------------------------------------------

if "jobDescription" in job_df.columns:

    empty_descriptions = (
        job_df["jobDescription"]
        .fillna("")
        .astype(str)
        .str.strip()
        .eq("")
        .sum()
    )

else:

    empty_descriptions = len(job_df)

add_test(
    "Job descriptions present",
    "G2-A",
    empty_descriptions == 0,
    f"Empty descriptions: {empty_descriptions}"
)

# ------------------------------------------------------------
# 10. COUNTRY TEST
# ------------------------------------------------------------

if "country" in job_df.columns:

    empty_country = (
        job_df["country"]
        .fillna("")
        .astype(str)
        .str.strip()
        .eq("")
        .sum()
    )

else:

    empty_country = len(job_df)

add_test(
    "Job country present",
    "G2-A",
    empty_country == 0,
    f"Empty country values: {empty_country}"
)

# ------------------------------------------------------------
# 11. RESUME ID TEST
# ------------------------------------------------------------

resume_id_valid = (
    "candidateId" in resume_df.columns
    and
    resume_df["candidateId"]
    .notna()
    .all()
    and
    resume_df["candidateId"]
    .astype(str)
    .str.strip()
    .ne("")
    .all()
)

add_test(
    "Candidate IDs are non-empty",
    "G2-B",
    resume_id_valid,
    f"Candidates checked: {len(resume_df)}"
)

# ------------------------------------------------------------
# 12. RESUME DUPLICATE TEST
# ------------------------------------------------------------

if "candidateId" in resume_df.columns:

    duplicate_candidates = (
        resume_df["candidateId"]
        .astype(str)
        .duplicated()
        .sum()
    )

else:

    duplicate_candidates = -1

add_test(
    "Duplicate Candidate IDs",
    "G2-B",
    duplicate_candidates == 0,
    f"Duplicate IDs: {duplicate_candidates}"
)

# ------------------------------------------------------------
# 13. RESUME TEXT TEST
# ------------------------------------------------------------

if "candidate_profile_text" in resume_df.columns:

    empty_resumes = (
        resume_df[
            "candidate_profile_text"
        ]
        .fillna("")
        .astype(str)
        .str.strip()
        .eq("")
        .sum()
    )

else:

    empty_resumes = len(resume_df)

add_test(
    "Resume profile text present",
    "G2-B",
    empty_resumes == 0,
    f"Empty resumes: {empty_resumes}"
)

# ------------------------------------------------------------
# 14. 30+ RESUME TEST
# ------------------------------------------------------------

resume_count_pass = (
    len(resume_df) >= 30
)

add_test(
    "At least 30 candidate profiles",
    "G2-B",
    resume_count_pass,
    f"Candidate records: {len(resume_df)}"
)

# ------------------------------------------------------------
# 15. VALIDATED BENCHMARK TEST
# ------------------------------------------------------------

if "benchmark_status" in resume_df.columns:

    benchmark_values = (
        resume_df[
            "benchmark_status"
        ]
        .astype(str)
        .str.upper()
    )

    validated_count = (
        benchmark_values
        .eq("MANUALLY_VALIDATED")
        .sum()
    )

    benchmark_pass = (
        validated_count == len(resume_df)
    )

else:

    validated_count = 0
    benchmark_pass = False

add_test(
    "Resume benchmark status",
    "G2-B",
    benchmark_pass,
    f"Manually validated records: "
    f"{validated_count}/{len(resume_df)}"
)

# ------------------------------------------------------------
# 16. SKILL LIST TEST
# ------------------------------------------------------------

def skill_count(value):

    if value is None:
        return 0

    if isinstance(
        value,
        (list, tuple, set)
    ):
        return len(value)

    text = str(value).strip()

    if not text or text.lower() in [
        "nan",
        "none",
        "[]"
    ]:
        return 0

    try:

        parsed = ast.literal_eval(
            text
        )

        if isinstance(
            parsed,
            (list, tuple, set)
        ):
            return len(parsed)

    except:
        pass

    return len([
        x
        for x in re.split(
            r"[,;|]",
            text
        )
        if x.strip()
    ])

# Job zero skills
if "normalized_skills" in job_df.columns:

    job_zero_skill = (
        job_df[
            "normalized_skills"
        ]
        .apply(skill_count)
        .eq(0)
        .sum()
    )

else:

    job_zero_skill = len(job_df)

job_zero_rate = (
    job_zero_skill / len(job_df)
    if len(job_df)
    else 0
)

# Resume zero skills
if "candidate_semantic_skills" in resume_df.columns:

    resume_zero_skill = (
        resume_df[
            "candidate_semantic_skills"
        ]
        .apply(skill_count)
        .eq(0)
        .sum()
    )

else:

    resume_zero_skill = len(resume_df)

resume_zero_rate = (
    resume_zero_skill / len(resume_df)
    if len(resume_df)
    else 0
)

add_test(
    "Job zero-skill measurement",
    "G2-A",
    True,
    f"{job_zero_skill}/{len(job_df)} "
    f"({job_zero_rate:.2%})"
)

add_test(
    "Resume zero-skill measurement",
    "G2-B",
    True,
    f"{resume_zero_skill}/{len(resume_df)} "
    f"({resume_zero_rate:.2%})"
)

# ------------------------------------------------------------
# 17. WEIGHTED GAP SCORE TEST
# ------------------------------------------------------------

if "weighted_gap_score" in resume_df.columns:

    gap_scores = pd.to_numeric(
        resume_df[
            "weighted_gap_score"
        ],
        errors="coerce"
    )

    invalid_gap_scores = (
        gap_scores.isna().sum()
    )

    out_of_range_gap = (
        (
            (gap_scores < 0)
            |
            (gap_scores > 1)
        )
        .fillna(False)
        .sum()
    )

    gap_pass = (
        invalid_gap_scores == 0
        and
        out_of_range_gap == 0
    )

else:

    invalid_gap_scores = len(resume_df)
    out_of_range_gap = len(resume_df)
    gap_pass = False

add_test(
    "Weighted gap score range",
    "G2-B",
    gap_pass,
    f"Invalid: {invalid_gap_scores}; "
    f"Out of range: {out_of_range_gap}"
)

# ------------------------------------------------------------
# 18. WEIGHTED GAP PERCENTAGE TEST
# ------------------------------------------------------------

if "weighted_gap_percentage" in resume_df.columns:

    gap_percentage = pd.to_numeric(
        resume_df[
            "weighted_gap_percentage"
        ],
        errors="coerce"
    )

    invalid_percentage = (
        gap_percentage.isna().sum()
    )

    out_of_range_percentage = (
        (
            (gap_percentage < 0)
            |
            (gap_percentage > 100)
        )
        .fillna(False)
        .sum()
    )

    percentage_pass = (
        invalid_percentage == 0
        and
        out_of_range_percentage == 0
    )

else:

    invalid_percentage = len(resume_df)
    out_of_range_percentage = len(resume_df)
    percentage_pass = False

add_test(
    "Weighted gap percentage range",
    "G2-B",
    percentage_pass,
    f"Invalid: {invalid_percentage}; "
    f"Out of range: {out_of_range_percentage}"
)

# ------------------------------------------------------------
# 19. PROCESSING METADATA TEST
# ------------------------------------------------------------

for dataset_name, data in [
    ("G2-A", job_df),
    ("G2-B", resume_df)
]:

    for column in [
        "nlp_model_version",
        "processing_version",
        "processing_date"
    ]:

        if column in data.columns:

            missing = (
                data[column]
                .isna()
                .sum()
            )

            add_test(
                f"{column} present",
                dataset_name,
                missing == 0,
                f"Missing values: {missing}"
            )

        else:

            add_test(
                f"{column} present",
                dataset_name,
                False,
                "Column missing"
            )

# ------------------------------------------------------------
# 20. FINAL STATUS
# ------------------------------------------------------------

quality_results_df = pd.DataFrame(
    test_results
)

total_tests = len(
    quality_results_df
)

passed_tests = (
    quality_results_df[
        "status"
    ]
    .eq("PASS")
    .sum()
)

failed_tests = (
    quality_results_df[
        "status"
    ]
    .eq("FAIL")
    .sum()
)

overall_status = (
    "PASS"
    if failed_tests == 0
    else "REVIEW_REQUIRED"
)

print("\n==============================================")
print("QUALITY TEST SUMMARY")
print("==============================================")

print(
    "Total tests:",
    total_tests
)

print(
    "Passed:",
    passed_tests
)

print(
    "Failed:",
    failed_tests
)

print(
    "Overall status:",
    overall_status
)

display(
    quality_results_df
)

# ------------------------------------------------------------
# 21. DATASET SUMMARY
# ------------------------------------------------------------

dataset_summary_df = pd.DataFrame({

    "dataset": [
        "G2-A Job Dataset",
        "G2-B Resume Dataset"
    ],

    "records": [
        len(job_df),
        len(resume_df)
    ],

    "unique_ids": [
        job_df["jobId"].nunique()
        if "jobId" in job_df.columns
        else 0,

        resume_df["candidateId"].nunique()
        if "candidateId" in resume_df.columns
        else 0
    ],

    "zero_skill_records": [
        job_zero_skill,
        resume_zero_skill
    ],

    "zero_skill_rate": [
        round(job_zero_rate, 4),
        round(resume_zero_rate, 4)
    ],

    "quality_status": [
        overall_status,
        overall_status
    ]

})

print("\nDATASET SUMMARY")
display(
    dataset_summary_df
)

# ------------------------------------------------------------
# 22. SAVE QUALITY REPORT
# ------------------------------------------------------------

output_file = (
    "45_FINAL_Data_Quality_Integration_Test.xlsx"
)

with pd.ExcelWriter(
    output_file,
    engine="openpyxl"
) as writer:

    quality_results_df.to_excel(
        writer,
        sheet_name="Quality_Tests",
        index=False
    )

    dataset_summary_df.to_excel(
        writer,
        sheet_name="Dataset_Summary",
        index=False
    )

print(
    "\n💾 Saved:",
    output_file
)

# ------------------------------------------------------------
# 23. DOWNLOAD
# ------------------------------------------------------------

files.download(
    output_file
)

print("\n==============================================")
print("✅ FINAL QUALITY TEST COMPLETED")
print("==============================================")

if overall_status == "PASS":

    print(
        "🎉 ALL QUALITY TESTS PASSED"
    )

else:

    print(
        "⚠️ SOME TESTS REQUIRE REVIEW."
    )

print("==============================================")
```
